**Notation:**
- $B$: batch size
- $L_1$: attending sequence length, $L_2$: being attended sequence length
- $D_1$: hidden dimension of attending sequence

- $W^V \in \mathbb{R}^{D_2 \times (H \cdot D_v)}$: concatenated value projection
- $W^O \in \mathbb{R}^{(H \cdot D_v) \times D_1}$: output projection
