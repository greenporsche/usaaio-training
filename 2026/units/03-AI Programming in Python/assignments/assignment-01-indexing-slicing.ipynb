## Context

NumPy's `ndarray` is the fundamental data structure in scientific Python and AI programming. Before you can perform any computation on data, you must be able to **select**, **extract**, and **rearrange** elements from arrays efficiently.

NumPy provides three indexing mechanisms:

- **Basic indexing** (slicing): uses `start:stop:step` notation, returns **views** (shared memory with the original array)
- **Fancy indexing**: uses integer arrays as indices, returns **copies**
- **Boolean indexing**: uses boolean arrays as masks, returns **copies**

### Notation

- Shape annotations follow every operation: `# (N, D)` means N rows and D columns
- We use the convention that `axis=0` runs along rows (downward) and `axis=1` runs along columns (across)
- A **view** shares memory with the original array; modifying a view modifies the original. A **copy** is independent.

This assignment tests your fluency with all three indexing mechanisms and your understanding of the view/copy distinction.

When you modify a view, you silently modify the original array. When you think you have a view but actually have a copy, your modifications are lost.
