## Context

Real-world AI data comes as structured tables -- CSV files, databases, spreadsheets. Pandas is the standard Python library for manipulating tabular data. Under the hood, each Pandas column is a NumPy array, so operations are fast.

### Key Concepts

- **DataFrame**: a 2D labeled table (rows have an index, columns have names)
- **Series**: a single column (1D labeled array)
- **GroupBy**: split data by a key, apply a function to each group, combine results
- **Merge**: join two tables on a shared key (like SQL JOIN)
- **Method chaining**: `df.query(...).assign(...).groupby(...).agg(...)` -- a pipeline of transformations
