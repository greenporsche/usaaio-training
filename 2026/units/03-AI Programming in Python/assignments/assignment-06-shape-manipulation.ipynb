## Context

Shape manipulation is the bridge between mathematical notation and executable code. When a paper says "reshape the input to a sequence of patches," you need to know exactly which NumPy operation does that -- and what happens to the data layout in memory.

USAAIO problems frequently involve converting data between formats: channels-last to channels-first, flat vectors to batched sequences, separate arrays into stacked tensors. Every shape error is a silent bug that produces wrong results without crashing.

### Key Operations

- **`reshape`**: changes shape without moving data (returns a view when possible)
- **`transpose` / `swapaxes`**: reorders dimensions (always returns a view)
- **`np.stack` / `np.concatenate`**: combines arrays along new or existing axes
- **`np.split`**: breaks an array into sub-arrays
- **`np.expand_dims` / `np.squeeze`**: adds or removes size-1 dimensions
- **`np.tile` / `np.repeat`**: duplicates data along axes

### Notation

- Shape annotations are **required** on every operation: `# (B, H, W, C) -> (B, C, H, W)`
- `axis=0` runs downward (rows), `axis=1` runs across (columns)
- A **view** shares memory; a **copy** is independent

Knowing which to use is critical for constructing positional encodings, checkerboard patterns, and repeated templates.
