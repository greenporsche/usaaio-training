## Context

Broadcasting is NumPy's mechanism for performing arithmetic operations on arrays of **different shapes**. It is the single most important concept for writing vectorized code -- and vectorized code is what USAAIO tests.

### Broadcasting Rules

When operating on two arrays, NumPy compares shapes **element-wise from the trailing (rightmost) dimensions**:

1. If arrays differ in number of dimensions, the smaller shape is **padded with 1s on the left**.
2. Arrays with size 1 in a dimension act as if they had the size of the larger array in that dimension (they are "stretched").
3. If sizes disagree and neither is 1, a `ValueError` is raised.

### Notation

- Shape annotations are required: `# (N, D) + (D,) -> (N, D)`
- `keepdims=True` preserves the collapsed dimension as size 1, which is often essential for correct broadcasting.
