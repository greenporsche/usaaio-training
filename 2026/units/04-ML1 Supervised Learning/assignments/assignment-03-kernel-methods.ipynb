## Background

The **kernel trick** lets us implicitly compute in high-dimensional (even infinite-dimensional) feature spaces by only computing pairwise inner products: $K(x_i, x_j) = \phi(x_i)^T\phi(x_j)$.

**Kernel Ridge Regression**: $\hat{\alpha} = (K + \lambda I)^{-1}y$, prediction: $\hat{y}(x_*) = k_*^T\alpha$ where $k_{*i} = K(x_*, x_i)$.

### Notation

| Symbol | Shape | Description |
|--------|-------|-------------|
| $K$ | $(n, n)$ | Kernel (Gram) matrix |
| $\alpha$ | $(n,)$ | Dual weights |
| $k_*$ | $(n,)$ | Kernel evaluations for test point |
| $\sigma$ | scalar | RBF bandwidth |
