## Background

When features are correlated or $d > n$, OLS overfits. **Regularization** penalizes the weights:

- **Ridge** (L2): $\;\mathcal{L} = \frac{1}{n}\|Xw - y\|^2 + \lambda\|w\|_2^2 \quad\Rightarrow\quad \hat{w} = (X^TX + \lambda I)^{-1}X^Ty$
- **Lasso** (L1): $\;\mathcal{L} = \frac{1}{n}\|Xw - y\|^2 + \lambda\|w\|_1$ — no closed form; use coordinate descent with **soft thresholding** $S_\lambda(x) = \text{sign}(x)\max(|x| - \lambda, 0)$

### Notation

| Symbol | Description |
|--------|-------------|
| $\lambda$ | Regularization strength |
| $\|w\|_2^2$ | $\sum_j w_j^2$ (L2 squared norm) |
| $\|w\|_1$ | $\sum_j |w_j|$ (L1 norm) |
