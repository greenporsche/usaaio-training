## Background

A **support vector machine** finds the maximum-margin hyperplane $w^Tx + b = 0$ separating two classes ($y \in \{-1, +1\}$). The margin is $\frac{2}{\|w\|}$.

**Soft-margin SVM** objective: $\min_{w,b}\; \frac{1}{2}\|w\|^2 + C \cdot \frac{1}{n}\sum_{i=1}^{n}\max(0,\; 1 - y_i(w^Tx_i + b))$

The **hinge loss** $\ell(z) = \max(0, 1-z)$ penalizes misclassifications and margin violations.

### Notation

| Symbol | Shape | Description |
|--------|-------|-------------|
| $w$ | $(d,)$ | Normal vector to hyperplane |
| $b$ | scalar | Bias (offset) |
| $C$ | scalar | Regularization trade-off |
| margin | scalar | $2 / \|w\|$ |

### The Role of $C$

The parameter $C$ controls the trade-off between maximizing the margin (small $\|w\|$) and minimizing hinge loss (correctly classifying training points). Large $C$ prioritizes correct classification (narrow margin, few violations). Small $C$ prioritizes a wide margin (more violations tolerated).
