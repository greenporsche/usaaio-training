## Background

**k-Nearest Neighbors** is a non-parametric method: to predict for a query $x_*$, find the $k$ training points closest to $x_*$ and aggregate their labels (majority vote for classification, mean for regression).

**Pairwise distances** can be computed without loops using: $\|a - b\|^2 = \|a\|^2 + \|b\|^2 - 2a^Tb$.

### Notation

| Symbol | Shape | Description |
|--------|-------|-------------|
| $X_{\text{train}}$ | $(n, d)$ | Training features |
| $X_{\text{test}}$ | $(m, d)$ | Test features |
| $D$ | $(m, n)$ | Pairwise distance matrix |
| $k$ | scalar | Number of neighbors |

### The Curse of Dimensionality

In high dimensions, distances between points **concentrate**: $\frac{d_{\max} - d_{\min}}{d_{\min}} \to 0$ as $d \to \infty$. This makes nearest-neighbor distinctions meaningless. kNN degrades when many features are irrelevant.
