```python
"""
DO NOT MAKE ANY CHANGE IN THIS CELL.
Numerical gradient verification for all four.
"""
def numerical_gradient(loss_fn, w, eps=1e-5):
    grad = np.zeros_like(w)
    for i in range(w.size):
        w_flat = w.ravel()
        w_plus = w_flat.copy(); w_plus[i] += eps
        w_minus = w_flat.copy(); w_minus[i] -= eps
        grad.ravel()[i] = (
            loss_fn(w_plus.reshape(w.shape)) - loss_fn(w_minus.reshape(w.shape))
        ) / (2 * eps)
    return grad

# Setup
X_g = np.random.randn(30, 4)                                     # (30, 4)
y_g_cont = np.random.randn(30)                                   # (30,)
y_g_bin = (np.random.rand(30) > 0.5).astype(float)              # (30,)
y_g_pm = 2 * y_g_bin - 1                                        # (30,) in {-1, +1}
Y_g_oh = np.zeros((30, 3)); Y_g_oh[np.arange(30), np.random.randint(0, 3, 30)] = 1
w_g = np.random.randn(4) * 0.1                                  # (4,)
W_g = np.random.randn(4, 3) * 0.1                               # (4, 3)

# Sigmoid helper (must be defined for logistic gradient to work)
def _sigmoid(z):
    return np.where(z >= 0, 1/(1+np.exp(-z)), np.exp(z)/(1+np.exp(z)))

def _softmax(Z):
    Z_shift = Z - Z.max(axis=1, keepdims=True)
    E = np.exp(Z_shift)
    return E / E.sum(axis=1, keepdims=True)

checks = [
    ("Ridge",
     ridge_gradient(X_g, y_g_cont, w_g, 0.5),
     numerical_gradient(lambda w: (1/30)*np.sum((X_g@w - y_g_cont)**2) + 0.5*np.sum(w**2), w_g)),
    ("Logistic",
     logistic_gradient(X_g, y_g_bin, w_g),
     numerical_gradient(lambda w: -(1/30)*np.sum(y_g_bin*np.log(_sigmoid(X_g@w)+1e-12) + (1-y_g_bin)*np.log(1-_sigmoid(X_g@w)+1e-12)), w_g)),
    ("Softmax",
     softmax_ce_gradient(X_g, Y_g_oh, W_g),
     numerical_gradient(lambda W: -(1/30)*np.sum(Y_g_oh*np.log(_softmax(X_g@W)+1e-12)), W_g)),
    ("Hinge",
     hinge_subgradient(X_g, y_g_pm, w_g),
     numerical_gradient(lambda w: (1/30)*np.sum(np.maximum(0, 1 - y_g_pm*(X_g@w))), w_g)),
]

for name, ana, num in checks:
    diff = np.max(np.abs(ana - num))
    status = "PASS" if diff < 1e-4 else "FAIL"
    print(f"{name:10s}: max|ana-num| = {diff:.2e}  [{status}]")
    assert diff < 1e-4, f"{name} gradient check failed"
print("Problem 4 passed.")
```
