**Notation:**
- Primal: $\min_{w,b} \frac{1}{2}\|w\|^2 + C\sum_i \xi_i$ s.t. $y_i(w^Tx_i + b) \geq 1 - \xi_i$

- Margin width: $\frac{2}{\|w\|}$
- Dual: $\max_\alpha \sum_i \alpha_i - \frac{1}{2}\sum_{i,j} \alpha_i \alpha_j y_i y_j K(x_i, x_j)$
