**Notation:**
- $X_c = X - \bar{X}$: centered data matrix, shape $(N, D)$
- $\Sigma = \frac{1}{N-1}X_c^T X_c$: covariance matrix, shape $(D, D)$
- $\Sigma v_k = \lambda_k v_k$: eigenvalue equation
- $Z = X_c V_k$: projected data, where $V_k = [v_1, \ldots, v_k]$
- Variance explained by component $k$: $\frac{\lambda_k}{\sum_j \lambda_j}$
