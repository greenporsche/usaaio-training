**Notation:**
- $X \in \mathbb{R}^{N \times D}$: feature matrix ($N$ samples, $D$ features)
- $y \in \{0, 1, \ldots, K-1\}^N$: class labels
- $p_k$: proportion of class $k$ in a node
- Gini impurity: $G = 1 - \sum_{k} p_k^2$
- Entropy: $H = -\sum_{k} p_k \log_2 p_k$
- Information gain: $IG = H_{\text{parent}} - \sum_v \frac{|S_v|}{|S|} H(S_v)$
