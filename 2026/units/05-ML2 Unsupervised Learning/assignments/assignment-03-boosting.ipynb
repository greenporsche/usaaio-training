**Notation:**
- $w_i^{(t)}$: weight of sample $i$ at round $t$
- $\epsilon_t = \sum_{i: h_t(x_i) \neq y_i} w_i^{(t)}$: weighted error of weak learner $h_t$
- $\alpha_t = \frac{1}{2}\ln\frac{1 - \epsilon_t}{\epsilon_t}$: learner weight
- $H(x) = \text{sign}\left(\sum_t \alpha_t h_t(x)\right)$: final ensemble
