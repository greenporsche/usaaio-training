## Background

Eigendecomposition is the factorization $A = Q\Lambda Q^{-1}$ where $\Lambda$ is diagonal with eigenvalues and columns of $Q$ are the corresponding eigenvectors. The **power method** iteratively computes the dominant eigenpair by repeatedly multiplying a random vector by $A$ and normalizing. **Deflation** removes a known eigenpair from the matrix so the power method can find the next one.

These techniques are foundational: Google's PageRank, PCA, Markov chain analysis, and spectral clustering all rely on eigendecomposition.

## Notation

| Symbol | Meaning |
|--------|---------|
| $A$ | square matrix, shape $(n, n)$ |
| $\lambda_i$ | $i$-th eigenvalue of $A$ |
| $\mathbf{v}_i$ | unit eigenvector for $\lambda_i$ |
| $Q$ | matrix of eigenvectors as columns, shape $(n, n)$ |
| $\Lambda$ | diagonal matrix of eigenvalues, shape $(n, n)$ |
