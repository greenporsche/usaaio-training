## Background

Principal Component Analysis (PCA) finds the directions of maximum variance in data. Given centered data $\bar{X} \in \mathbb{R}^{N \times d}$, the covariance matrix $C = \frac{1}{N-1}\bar{X}^\top\bar{X}$ captures the variance structure. PCA computes the eigendecomposition of $C$ and projects onto the top-$k$ eigenvectors. Equivalently, PCA can be computed via SVD of the centered data matrix, which is numerically more stable and avoids forming $C$ explicitly.

## Notation

| Symbol | Meaning |
|--------|---------|
| $X$ | data matrix, shape $(N, d)$ — $N$ samples, $d$ features |
| $\bar{X}$ | centered data: $X - \mathbf{1}\boldsymbol{\mu}^\top$ |
| $C$ | covariance matrix, shape $(d, d)$ |
| $W_k$ | top-$k$ eigenvectors of $C$, shape $(d, k)$ |
| $Z$ | projected scores $\bar{X}W_k$, shape $(N, k)$ |
| $P$ | projection matrix $W_k W_k^\top$, shape $(d, d)$ |
