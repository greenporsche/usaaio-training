## Background

A quadratic form $Q(\mathbf{x}) = \mathbf{x}^\top A \mathbf{x}$ maps vectors to scalars. Its sign behavior determines the classification of $A$: positive definite (PD), positive semi-definite (PSD), negative definite, negative semi-definite, or indefinite. PD matrices arise everywhere: covariance matrices are PSD, Hessians at local minima are PSD, and kernel matrices must be PSD. Cholesky decomposition $A = LL^\top$ is both a test for PD-ness and a fast way to solve linear systems.

## Notation

| Symbol | Meaning |
|--------|---------|
| $Q(\mathbf{x})$ | quadratic form $\mathbf{x}^\top A \mathbf{x}$ |
| PD / PSD | positive definite / positive semi-definite |
| $L$ | lower triangular Cholesky factor |
| $d_M$ | Mahalanobis distance |
