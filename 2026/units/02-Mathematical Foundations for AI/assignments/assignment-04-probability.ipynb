## Background

Probability theory underpins nearly every ML algorithm. Bayesian inference provides a principled framework for updating beliefs given data: $P(\theta|D) \propto P(D|\theta)P(\theta)$. Maximum Likelihood Estimation (MLE) finds the parameters that maximize $P(D|\theta)$. The multivariate Gaussian distribution is the most important continuous distribution in ML, parameterized by mean $\boldsymbol{\mu}$ and covariance $\Sigma$.

## Notation

| Symbol | Meaning |
|--------|---------|
| $\text{Beta}(\alpha, \beta)$ | Beta distribution with shape parameters $\alpha, \beta > 0$ |
| $\mathcal{N}(\boldsymbol{\mu}, \Sigma)$ | multivariate Gaussian |
| $\hat{\mu}, \hat{\Sigma}$ | MLE parameter estimates |
| $\|\Sigma\|$ | determinant of $\Sigma$ |

Maximum Likelihood Estimation finds $\hat{\theta} = \arg\max_\theta \prod_{i=1}^N p(x_i | \theta)$. For Gaussians, the MLE for the mean is $\hat{\mu} = \frac{1}{N}\sum x_i$ and for variance is $\hat{\sigma}^2 = \frac{1}{N}\sum(x_i - \hat{\mu})^2$. Note: this is the **biased** variance estimator.

The MLE variance estimator $\hat{\sigma}^2 = \frac{1}{N}\sum(x_i - \bar{x})^2$ is biased: $\mathbb{E}[\hat{\sigma}^2] = \frac{N-1}{N}\sigma^2$. This is because $\bar{x}$ itself uses one degree of freedom from the data. The unbiased estimator divides by $N-1$ instead.
