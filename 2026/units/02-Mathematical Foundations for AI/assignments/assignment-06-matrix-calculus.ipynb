## Background

Matrix calculus is the language of deep learning. Every parameter update requires computing gradients of scalar loss functions with respect to vectors and matrices. The chain rule generalizes to Jacobians: if $L = g(f(\mathbf{x}))$, then $\frac{\partial L}{\partial \mathbf{x}} = \frac{\partial g}{\partial f} \frac{\partial f}{\partial \mathbf{x}}$. Backpropagation applies this chain rule layer by layer. Numerical gradient checking validates analytical gradients.

## Notation

| Symbol | Meaning |
|--------|---------|
| $\nabla_{\mathbf{x}} f$ | gradient of scalar $f$ w.r.t. vector $\mathbf{x}$ |
| $\frac{\partial L}{\partial W}$ | gradient of loss w.r.t. weight matrix |
| $\mathbf{e}_y$ | one-hot vector with 1 at index $y$ |
| $\text{softmax}(\mathbf{z})_k$ | $e^{z_k}/\sum_j e^{z_j}$ |

The following numerical gradient checker uses central differences to approximate the gradient. Use it to verify every analytical gradient you derive.

```python
# Provided helper — do not modify
def numerical_gradient(f, x: np.ndarray, eps: float = 1e-5) -> np.ndarray:
    """Central difference numerical gradient."""
    grad = np.zeros_like(x)
    it = np.nditer(x, flags=['multi_index'], op_flags=['readwrite'])
    while not it.finished:
        idx = it.multi_index
        old_val = x[idx]
        x[idx] = old_val + eps
        f_plus = f(x)
        x[idx] = old_val - eps
        f_minus = f(x)
        grad[idx] = (f_plus - f_minus) / (2 * eps)
        x[idx] = old_val
        it.iternext()
    return grad

def check_gradient(f, grad_f, x, name=""):
    """Compare analytical and numerical gradients."""
    g_analytical = grad_f(x)
    g_numerical = numerical_gradient(f, x.copy())
    diff = np.linalg.norm(g_analytical - g_numerical) / (
        np.linalg.norm(g_analytical) + np.linalg.norm(g_numerical) + 1e-15)
    status = "PASS" if diff < 1e-5 else "FAIL"
    print(f"{name}: relative diff = {diff:.2e} [{status}]")
    return diff < 1e-5
```
