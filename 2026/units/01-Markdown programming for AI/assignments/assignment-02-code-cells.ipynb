## Background

A well-organized notebook interleaves **markdown cells** (explanations, formulas, formatted answers) with **code cells** (computations, data processing, verification). On USAAIO, the grader reads your notebook top-to-bottom. If the structure is unclear — if the grader cannot tell which code answers which question, or which markdown cell states which result — you lose points even when the computation is correct.

This assignment tests your ability to organize notebooks using the standard USAAIO pattern:

1. **Markdown**: State the problem, show the formula
2. **Code**: Compute the answer
3. **Markdown**: State the result in bold, with interpretation

## Notation
