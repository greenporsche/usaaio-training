A single neuron computes $\sigma(w^T x + b)$, which partitions the input space with a **half-plane**. The set $H^+ = \{x : w^T x + b > 0\}$ is the positive side. An MLP stacks multiple neurons to combine these half-planes using Boolean-like logic, forming arbitrarily complex decision boundaries. In this problem, you will analyze how MLPs carve up 2D space to solve classification problems, including the classic XOR function and the parity function.

We use the following notation in this problem.
- $x \in \mathbb{R}^2$ — 2D input point.
- $w \in \mathbb{R}^2$, $b \in \mathbb{R}$ — weight vector and bias for a single neuron.
- $\mathbb{1}[\cdot]$ — indicator (step) function: returns 1 if the argument is true, 0 otherwise.
- $H^+(w, b) = \{x : w^T x + b > 0\}$ — the positive half-plane defined by neuron $(w, b)$.
- A **decision boundary** is the set $\{x : w^T x + b = 0\}$.
- Architecture notation: $d_0 \to d_1 \to \cdots \to d_L$ means input dimension $d_0$, hidden layers $d_1, \ldots, d_{L-1}$, output dimension $d_L$.
