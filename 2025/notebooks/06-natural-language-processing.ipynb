### 1.2 Text Representation Challenges

One of the fundamental challenges in NLP is how to represent text in a way that computers can process effectively. Humans intuitively understand language, but computers operate on numbers, not words.

#### The Representation Problem

The **representation problem** in NLP refers to the challenge of converting human language (symbolic, discrete, and ambiguous) into numerical forms that machines can process while preserving meaning.

Consider these challenges:

1. **Discreteness**: Text consists of discrete symbols (characters, words) while machine learning often works best with continuous values
2. **High-dimensionality**: Language has a large vocabulary - how do we represent thousands or millions of words efficiently?
3. **Sparsity**: Any single document uses only a small fraction of possible words
4. **Semantic gap**: Similar meanings can be expressed with entirely different words

#### Ambiguity in Language

Natural language is inherently ambiguous at multiple levels:

- Lexical ambiguity: Words with multiple meanings ("bank" can mean financial institution or river edge)
- Syntactic ambiguity: Sentences with multiple grammatical interpretations ("I saw the person with binoculars")
- Referential ambiguity: Unclear pronoun references ("John told Bill that he was late")

````python
### Demonstrating the Curse of Dimensionality in Text Data

import numpy as np
import matplotlib.pyplot as plt
from sklearn.feature_extraction.text import CountVectorizer
import pandas as pd

# Example corpus
corpus = [
    "This is the first document.",
    "This document is the second document.",
    "And this is the third one.",
    "Is this the first document?"
]

# Create a Bag of Words representation
vectorizer = CountVectorizer()
X = vectorizer.fit_transform(corpus)

# Get the feature names (words)
feature_names = vectorizer.get_feature_names_out()

# Convert to a DataFrame for better visualization
df = pd.DataFrame(X.toarray(), columns=feature_names)
print("Bag of Words representation:")
print(df)
print(f"\nShape of the matrix: {X.shape} (documents × vocabulary size)")
print(f"Sparsity: {(X.toarray() == 0).sum() / (X.shape[0] * X.shape[1]) * 100:.2f}% of the matrix is zeros")

# Visualize the sparsity
plt.figure(figsize=(10, 6))
plt.spy(X, markersize=10, color='blue')
plt.title("Sparsity in Bag of Words Matrix")
plt.xlabel("Word Index")
plt.ylabel("Document Index")
plt.tight_layout()
plt.show()

# This demonstrates how quickly the vocabulary grows with more text
word_counts = []
doc_counts = [1, 2, 3, 4]
for i in doc_counts:
    vec = CountVectorizer()
    vec.fit(corpus[:i])
    word_counts.append(len(vec.vocabulary_))

plt.figure(figsize=(10, 6))
plt.plot(doc_counts, word_counts, marker='o')
plt.title("Vocabulary Growth with Document Count")
plt.xlabel("Number of Documents")
plt.ylabel("Vocabulary Size")
plt.grid(True)
plt.tight_layout()
plt.show()

print("\nThis demonstrates the 'curse of dimensionality' in text processing:")
print("1. As we add more documents, the vocabulary size grows")
print("2. The matrix becomes increasingly sparse (mostly zeros)")
print("3. Most words appear in only a small fraction of documents")
print("4. This makes computing similarities challenging")
````

### 1.3 From Characters to Meaning: The NLP Pipeline

To process natural language, we typically follow a series of steps that transform raw text into structured representations that machines can use for various tasks. This sequence of transformations is often called the **NLP pipeline**.

#### The Core NLP Pipeline

1. **Text Cleaning**: Remove unwanted characters, normalize text
2. **Tokenization**: Break text into smaller units (characters, words, or subwords)
3. **Normalization**: Convert to lowercase, stem or lemmatize words
4. **Representation**: Convert tokens to numerical format (one-hot encoding, embeddings)
5. **Feature Engineering/Learning**: Create or learn features for the specific task
6. **Model Application**: Apply these representations to a downstream task


Each stage transforms language to bring it closer to a form that models can effectively process.

#### Importance of Tokenization and Embeddings

In this notebook, we'll focus especially on two crucial parts of the pipeline:

1. **Tokenization**: How we break text into meaningful units
2. **Embeddings**: How we convert these tokens into numerical vectors


These two steps form the foundation of almost all modern NLP systems. The choices we make at these stages dramatically affect the performance of downstream tasks. As we'll see in later sections, there's a rich variety of approaches to both tokenization and embedding, each with different strengths and weaknesses.

````python
### Exploring a Simple NLP Pipeline

# Let's see a simple version of an NLP pipeline in action
import re
import nltk
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer

# Download required NLTK resources (only needed first time)
try:
    nltk.data.find('tokenizers/punkt')
    nltk.data.find('corpora/stopwords')
except LookupError:
    nltk.download('punkt')
    nltk.download('stopwords')

# Sample text
text = "Natural Language Processing (NLP) is a field of AI that focuses on the interaction between computers and humans using natural language. The ultimate goal of NLP is to enable computers to understand, interpret, and generate human language in a valuable way."

# Step 1: Text cleaning
def clean_text(text):
    # Remove special characters and digits
    text = re.sub(r'[^a-zA-Z\s]', '', text)
    return text

# Step 2: Tokenization
def tokenize_text(text):
    return word_tokenize(text)

# Step 3: Normalization (lowercase, remove stopwords, stemming)
def normalize_tokens(tokens):
    # Convert to lowercase
    tokens = [token.lower() for token in tokens]
    
    # Remove stopwords
    stop_words = set(stopwords.words('english'))
    tokens = [token for token in tokens if token not in stop_words]
    
    # Stemming
    stemmer = PorterStemmer()
    tokens = [stemmer.stem(token) for token in tokens]
    
    return tokens

# Step 4: Simple representation (Bag of Words)
def create_bow(tokens):
    bow = {}
    for token in tokens:
        if token in bow:
            bow[token] += 1
        else:
            bow[token] = 1
    return bow

# Apply the pipeline steps
print(f"Original text:\n{text}\n")

cleaned_text = clean_text(text)
print(f"After cleaning:\n{cleaned_text}\n")

tokens = tokenize_text(cleaned_text)
print(f"After tokenization:\n{tokens}\n")

normalized_tokens = normalize_tokens(tokens)
print(f"After normalization:\n{normalized_tokens}\n")

bow = create_bow(normalized_tokens)
print(f"Bag of Words representation:\n{bow}")

# In modern NLP, we would replace the BoW with embeddings
print("\nNote: In upcoming sections, we'll replace this simple Bag of Words")
print("representation with much more powerful embedding techniques!")
````

### Aside: The Pre-neural Era

Before neural networks took over NLP, statistical approaches dominated the field. These methods, popular from the mid-1990s through the early 2010s, established many concepts we still use today.

**Statistical Machine Translation (SMT)** broke translation into three components: translation model (word-level correspondences), language model (fluency), and decoder (search). IBM's models I through V gradually increased sophistication, treating translation as a noisy channel problem. Google Translate originally used SMT before switching to neural methods in 2016.

**Hidden Markov Models (HMMs)** powered part-of-speech tagging and name entity recognition, treating language as a sequence of hidden states generating observable words. Their ability to capture local sequential dependencies made them well-suited for these tasks.

**Conditional Random Fields (CRFs)** extended sequence modeling by considering features from both past and future context, achieving state-of-the-art results for tasks like chunking and entity extraction.

**Feature engineering** was an art form in this era. Researchers carefully crafted features like surrounding words, affixes, capitalization patterns, and dictionary lookups. A good feature set could make or break a system.

**Noisy Channel Models** treated many NLP tasks as recovering an original "clean" signal from a corrupted version - applying this to spelling correction, text classification, and machine translation.

These approaches required linguistic expertise and careful feature design but laid important groundwork for understanding language statistically. The neural revolution didn't invalidate these insights - it automated feature learning while building on many of the same probabilistic foundations.

### 2.1 Tokenization Fundamentals

Tokenization is the process of breaking down text into smaller units called *tokens*. This might seem like a simple task - just split text by spaces, right? But as we'll discover, it's surprisingly complex and full of nuances.

#### What is a token?
A token is the smallest unit of text that we process in NLP. Depending on our approach, tokens could be:
* Individual characters
* Words
* Subword units (parts of words)
* Even phrases or sentences in some applications

#### Why tokenize?

Computers can't directly process raw text - they need numerical representations. Tokenization is the first step in converting text into these representations. It serves several critical functions:

1. Creates discrete units that can be counted, indexed, and processed
2. Establishes the vocabulary size for our NLP system
3. Determines the granularity of our text representation
4. Affects how much meaning each individual unit carries

#### The tokenization spectrum

There's a fundamental trade-off in tokenization between vocabulary size and sequence length:

* **Fine-grained tokenization** (e.g., characters): Small vocabulary, long sequences
* **Coarse-grained tokenization** (e.g., words): Large vocabulary, short sequences

![Tokenization Spectrum](https://stanford.edu/~shervine/teaching/cs-221/illustrations/token-word-character-sentence.png)
*The tokenization spectrum from characters to words to sentences, with examples of each approach.*

````python
### Basic tokenization example

# Let's start with a simple example showing different tokenization approaches
text = "The quick brown fox jumps over the lazy dog."

# Character tokenization
char_tokens = list(text)

# Word tokenization (simplistic approach using split)
word_tokens = text.split()

# Word tokenization with punctuation handling
import re
word_tokens_with_punct = re.findall(r'\w+|[^\w\s]', text)

print(f"Original text: {text}")
print(f"\nCharacter tokens: {char_tokens}")
print(f"Token count: {len(char_tokens)}")
print(f"Vocabulary size: {len(set(char_tokens))}")

print(f"\nSimple word tokens: {word_tokens}")
print(f"Token count: {len(word_tokens)}")
print(f"Vocabulary size: {len(set(word_tokens))}")

print(f"\nWord tokens with punctuation: {word_tokens_with_punct}")
print(f"Token count: {len(word_tokens_with_punct)}")
print(f"Vocabulary size: {len(set(word_tokens_with_punct))}")

# Visualize the sequence length vs vocabulary size trade-off
import matplotlib.pyplot as plt

token_types = ['Character', 'Word', 'Word+Punct']
token_counts = [len(char_tokens), len(word_tokens), len(word_tokens_with_punct)]
vocab_sizes = [len(set(char_tokens)), len(set(word_tokens)), len(set(word_tokens_with_punct))]

fig, ax = plt.subplots(figsize=(10, 5))
x = range(len(token_types))
width = 0.35

ax.bar([i - width/2 for i in x], token_counts, width, label='Sequence Length')
ax.bar([i + width/2 for i in x], vocab_sizes, width, label='Vocabulary Size')

ax.set_ylabel('Count')
ax.set_title('Tokenization Trade-off: Sequence Length vs. Vocabulary Size')
ax.set_xticks(x)
ax.set_xticklabels(token_types)
ax.legend()

plt.show()
````

#### Tokenization as a mathematical function

Formally, we can define tokenization as a function $T$ that maps a text string $S$ to a sequence of tokens $[t_1, t_2, ..., t_n]$:

$$T: S \rightarrow [t_1, t_2, ..., t_n]$$

An ideal tokenization function would have these properties:

1. **Invertible**: You should be able to reconstruct the original text from the tokens
2. **Meaningful**: Tokens should capture linguistically meaningful units
3. **Efficient**: The vocabulary size should be manageable
4. **Complete**: Any text should be tokenizable without unknown tokens

Unfortunately, no single tokenization approach satisfies all these properties perfectly, which is why we have multiple methods, each with their own trade-offs.

#### Vocabulary and coverage

The set of all possible tokens is called the *vocabulary* or *lexicon*. A key metric for tokenization methods is *coverage* - what percentage of text can be represented using the vocabulary?

For character tokenization, coverage is typically 100% (assuming a fixed character set like ASCII or Unicode). For word tokenization, coverage depends on the vocabulary size and the domain of text.

### Aside: The Vocabulary Explosion Problem

One of the most fundamental challenges in NLP is managing vocabulary size. Languages naturally have a long tail distribution of word frequencies - a small number of words appear very frequently, while a vast number appear rarely.

This creates what we call the *vocabulary explosion problem*. Let's put some numbers to it:

- English has about 170,000 words in current use according to the Oxford English Dictionary
- But when we include inflected forms (runs, running), proper names, technical terms, and other variations, we easily reach millions of unique word forms
- Social media introduces creative spellings, hashtags, and neologisms daily
- Technical fields have their own specialized terminology
- New words enter languages constantly

In the early days of NLP, systems would typically limit vocabularies to the 50,000-100,000 most frequent words and replace all other words with a special <UNK> (unknown) token. This was deeply unsatisfying - imagine reading a medical text where all the important technical terms are replaced with <UNK>!

This vocabulary problem was one of the main drivers behind the development of subword tokenization methods, which we'll explore later in this section.

### 2.2 Character Tokenization

Character tokenization is the simplest approach: we break text into individual characters. Each character becomes a token.

#### How it works

1. Iterate through each character in the text
2. Treat each character as a separate token
3. Optionally include spaces and punctuation as tokens

#### Advantages

- **Fixed, small vocabulary**: Typically just ASCII (128 characters) or Unicode (though Unicode has over 140,000 characters, most languages use a small subset)
- **No out-of-vocabulary problem**: Any text can be represented with the character set
- **Captures morphological patterns**: Prefixes, suffixes, and stems emerge naturally
- **Language-agnostic**: Works for any language without modification

#### Limitations

- **Very long sequences**: A typical sentence produces 5-10x more tokens than word-level tokenization
- **Limited semantic meaning**: Individual characters often carry no semantic information
- **Computational inefficiency**: Processing long sequences requires more memory and computation
- **Challenging for sequence models**: Harder to capture long-range dependencies

![Character Tokenization](https://miro.medium.com/max/1400/1*EOautM-5T_AaAqTKKVgBhA.png)
*Example of character tokenization, breaking text into individual characters.*

````python
### Character tokenization implementation

def character_tokenize(text):
    """
    Simple character tokenizer that preserves all characters including spaces and punctuation.
    """
    return list(text)

# Example with multiple languages
texts = [
    "Hello, world!",  # English
    "你好，世界！",     # Chinese
    "こんにちは、世界！"  # Japanese
]

for text in texts:
    tokens = character_tokenize(text)
    print(f"Original: {text}")
    print(f"Tokenized: {tokens}")
    print(f"Token count: {len(tokens)}")
    print(f"Vocabulary: {set(tokens)}")
    print(f"Vocabulary size: {len(set(tokens))}\n")

# Unicode handling is important for character tokenization
import unicodedata

def analyze_unicode(text):
    """
    Analyze the Unicode properties of each character in text
    """
    analysis = []
    for char in text:
        char_name = unicodedata.name(char, "Unknown")
        char_category = unicodedata.category(char)
        char_code = ord(char)
        analysis.append(f"{char}: U+{char_code:04X} {char_category} {char_name}")
    return analysis

# Analyze the first text from each language
for text in [texts[0][:5], texts[1][:2], texts[2][:3]]:
    print(f"Unicode analysis for '{text}':")
    for line in analyze_unicode(text):
        print(line)
    print()
````

#### Unicode and character encodings

When using character tokenization, it's essential to understand character encodings, particularly Unicode. Unicode provides a unique number for every character across all languages and scripts.

Some important concepts when dealing with character tokenization:

1. **Code points**: The unique numerical values assigned to each Unicode character
2. **Encodings**: Methods to represent code points as bytes (UTF-8, UTF-16, etc.)
3. **Normalization**: Converting equivalent characters to a consistent representation
4. **Combining characters**: Some characters are composed of multiple Unicode code points

For example, the character "é" can be represented as a single code point (U+00E9) or as the letter "e" followed by the combining acute accent (U+0065 + U+0301).

Character tokenization works best when text is properly normalized to avoid these ambiguities.

### Aside: Before Unicode

Before Unicode standardization in the 1990s, text processing was chaotic. Different operating systems, countries, and applications used incompatible encodings:

- ASCII covered just 128 characters, sufficient only for English with no accents
- EBCDIC was IBM's alternative to ASCII used on mainframes
- ISO-8859 had different variants for different European languages
- Shift-JIS, EUC-JP, and others were used for Japanese
- GB standards served Chinese users
- KOI8-R covered Russian

Transferring text between systems often resulted in *mojibake* - garbled characters that rendered text unreadable. A French letter with accents might display correctly on a French computer but become corrupted on a German system using a different encoding.

Early NLP systems were largely restricted to English and ASCII, with custom solutions needed for other languages. The ASCII-centrism of this era still echoes in programming languages and systems today, where English-centric assumptions are baked into tools and libraries.

Unicode's development represented one of computing's most important internationalization efforts. It unified previously fragmented character sets into a single standard that could represent all human writing systems. This laid the groundwork for truly multilingual NLP systems and enabled the global internet as we know it today.

When we use character tokenization with Unicode, we're building on this historical foundation - a remarkable achievement that makes it possible to process text in any language consistently.

### 2.3 Word Tokenization

Word tokenization splits text into word units, which feels intuitive for humans since words are meaningful semantic units in most languages.

#### How it works

There are several approaches to word tokenization:

1. **Simple whitespace splitting**: Split text at spaces
2. **Rule-based tokenization**: Apply language-specific rules for handling punctuation, contractions, etc.
3. **Statistical tokenization**: Use machine learning models to determine word boundaries
4. **Dictionary-based approaches**: Match text against a dictionary of known words

#### Challenges in word tokenization

Word tokenization is surprisingly challenging due to various linguistic phenomena:

- **Punctuation**: Should "end." be one token or two? How about "U.S.A."?
- **Contractions**: Is "don't" one token or two ("do" and "n't")?
- **Hyphenated words**: Should "state-of-the-art" be one token or multiple?
- **Compounds**: Is "icecream" one word or two?
- **Special tokens**: How to handle mentions (@username), hashtags (#trending), URLs, emoji?

#### Stemming and lemmatization

Often, word tokenization is combined with normalization techniques to reduce the vocabulary size and treat similar word forms as the same token:

- **Stemming** is a rule-based approach that chops off word endings to get an approximate common root (stem)
- **Lemmatization** uses linguistic knowledge to convert words to their dictionary form (lemma)

For example:
- Stemming: "running", "runner", and "runs" might all become "run"
- Lemmatization: "better" becomes "good", "was" becomes "be"

Here's how they differ:

| Aspect | Stemming | Lemmatization |
|--------|----------|---------------|
| Method | Rule-based | Dictionary/linguistic knowledge |
| Output | Often not a real word | Valid dictionary word |
| Speed | Fast | Slower |
| Accuracy | Lower | Higher |
| Example | "studies" → "studi" | "studies" → "study" |

*Comparison of stemming and lemmatization.*

````python
### Stemming and lemmatization examples

from nltk.stem import PorterStemmer, SnowballStemmer
from nltk.stem import WordNetLemmatizer
from nltk.corpus import wordnet

# Download necessary NLTK data
nltk.download('wordnet', quiet=True)
nltk.download('omw-1.4', quiet=True)

# Initialize stemmers and lemmatizer
porter = PorterStemmer()
snowball = SnowballStemmer("english")
lemmatizer = WordNetLemmatizer()

# Function to apply stemming and lemmatization
def normalize_words(words):
    results = {"Original": words}
    
    # Apply Porter stemmer
    results["Porter Stemmer"] = [porter.stem(word) for word in words]
    
    # Apply Snowball stemmer
    results["Snowball Stemmer"] = [snowball.stem(word) for word in words]
    
    # Apply lemmatizer - note that WordNet lemmatizer needs POS tag for best results
    # Here we just use noun as default for simplicity
    results["WordNet Lemmatizer"] = [lemmatizer.lemmatize(word) for word in words]
    
    return results

# Test with various words
test_words = [
    "running", "ran", "runs", 
    "better", "best", "good", 
    "studies", "studying", "studied",
    "mice", "cacti", "wolves",
    "meeting", "agreed", "fantastic"
]

results = normalize_words(test_words)

# Display results in a table
import pandas as pd

df = pd.DataFrame(results)
print(df)

# Let's try lemmatization with the correct part-of-speech
def lemmatize_with_pos(word, pos):
    if pos == 'v':
        return lemmatizer.lemmatize(word, wordnet.VERB)
    elif pos == 'n':
        return lemmatizer.lemmatize(word, wordnet.NOUN)
    elif pos == 'a':
        return lemmatizer.lemmatize(word, wordnet.ADJ)
    elif pos == 'r':
        return lemmatizer.lemmatize(word, wordnet.ADV)
    else:
        return lemmatizer.lemmatize(word)

pos_examples = [
    ("running", "v"),  # verb
    ("running", "n"),  # noun (as in "a running") 
    ("better", "a"),   # adjective
    ("better", "v"),   # verb
    ("well", "r")      # adverb
]

print("\nLemmatization with part-of-speech:")
for word, pos in pos_examples:
    print(f"{word} ({pos}) → {lemmatize_with_pos(word, pos)}")
````

### Aside: The Penn Treebank Standard

Much of English NLP tokenization follows conventions established by the Penn Treebank in the 1990s. The Penn Treebank was a groundbreaking project that manually annotated over 4.5 million words of American English with syntactic and semantic information.

The Penn Treebank tokenization standard includes these specific rules:

- Separating contractions: "don't" becomes "do" and "n't"
- Treating punctuation as separate tokens: "word." becomes "word" and "."
- Keeping hyphenated words together
- Splitting possessives: "John's" becomes "John" and "'s"

These decisions might seem arbitrary, but they've become standard in research benchmarks and evaluation metrics. For decades, NLP models were evaluated on Penn Treebank-tokenized data, so systems had to adopt these conventions to be comparable with published results.

The Penn Treebank's influence extended far beyond tokenization - its part-of-speech tag set and syntactic annotation style became de facto standards. Even today, when we've moved beyond many of its limitations, understanding the Penn Treebank conventions remains important for interpreting research results and legacy systems.

This illustrates an important point about NLP: seemingly minor preprocessing decisions can become deeply embedded in the field's practices and evaluations, influencing research directions for decades.

### 2.4 Subword Tokenization

Subword tokenization emerged as a middle ground between character and word tokenization, addressing limitations of both approaches. This family of methods breaks words into meaningful subunits like morphemes, syllables, or frequent character sequences.

#### The need for subword tokenization

Subword tokenization addresses several key problems:

1. **Out-of-vocabulary (OOV) words**: By breaking unknown words into known subwords, we can represent any word
2. **Vocabulary management**: Maintains a reasonable vocabulary size while preserving word meaning
3. **Morphologically rich languages**: Captures structure in languages with complex word formation
4. **Rare words**: Represents rare words effectively using common subunits

#### Common subword tokenization algorithms

Several subword tokenization methods have been developed:

1. **Byte Pair Encoding (BPE)**: Iteratively merges most frequent pairs of characters or character sequences
2. **WordPiece**: Similar to BPE but uses likelihood rather than frequency for merges
3. **SentencePiece**: Language-agnostic subword tokenization without pre-tokenization
4. **Unigram Language Model**: Probabilistic approach that optimizes a unigram language model

#### Byte Pair Encoding (BPE)

Byte Pair Encoding (BPE) was originally a data compression algorithm adapted for NLP by Sennrich et al. in 2016. It works as follows:

1. Start with a vocabulary of individual characters
2. Count all pairs of adjacent symbols in the training corpus
3. Replace the most frequent pair with a new symbol
4. Repeat steps 2-3 for a predetermined number of merges

Through this process, BPE learns a vocabulary of subword units that efficiently represents the training corpus.

#### WordPiece

WordPiece is similar to BPE but uses a different criterion for merges. Instead of choosing the most frequent pair, it chooses the pair that maximizes the likelihood of the training data when added to the vocabulary. It's used in models like BERT.

In the WordPiece representation, subword continuations are often marked with ## to indicate they're part of a larger word. For example, "playing" becomes ["play", "##ing"].

#### SentencePiece

SentencePiece is language-independent and treats the input as a raw Unicode string without any pre-tokenization. This makes it effective for languages without explicit word boundaries. It can use either BPE or Unigram Language Model algorithms internally.

````python
### Subword tokenization with Hugging Face transformers

# Install the transformers library if not already present
!pip install transformers -q

from transformers import BertTokenizer, GPT2Tokenizer, T5Tokenizer

# Load pre-trained tokenizers that use different subword approaches
bert_tokenizer = BertTokenizer.from_pretrained('bert-base-uncased')  # WordPiece
gpt2_tokenizer = GPT2Tokenizer.from_pretrained('gpt2')               # BPE
t5_tokenizer = T5Tokenizer.from_pretrained('t5-small')               # SentencePiece

# Test sentences including rare/complex words
test_sentences = [
    "The quick brown fox jumps over the lazy dog.",
    "I love natural language processing and deep learning!",
    "Supercalifragilisticexpialidocious is a very long word.",
    "In 2023, the latest NLP models can understand text remarkably well.",
    "Neural networks use backpropagation to optimize weights."
]

# Compare different tokenizers on the test sentences
def compare_subword_tokenizers(sentence):
    print(f"Original sentence: {sentence}\n")
    
    # BERT (WordPiece)
    bert_tokens = bert_tokenizer.tokenize(sentence)
    print(f"BERT (WordPiece): {bert_tokens}")
    print(f"Token count: {len(bert_tokens)}\n")
    
    # GPT-2 (BPE)
    gpt2_tokens = gpt2_tokenizer.tokenize(sentence)
    print(f"GPT-2 (BPE): {gpt2_tokens}")
    print(f"Token count: {len(gpt2_tokens)}\n")
    
    # T5 (SentencePiece)
    t5_tokens = t5_tokenizer.tokenize(sentence)
    print(f"T5 (SentencePiece): {t5_tokens}")
    print(f"Token count: {len(t5_tokens)}\n")
    print("-" * 80)

# Let's examine tokenization of a complex word in detail
complex_word = "Supercalifragilisticexpialidocious"
print(f"Complex word: {complex_word}\n")

bert_tokens = bert_tokenizer.tokenize(complex_word)
print(f"BERT (WordPiece): {bert_tokens}")

gpt2_tokens = gpt2_tokenizer.tokenize(complex_word)
print(f"GPT-2 (BPE): {gpt2_tokens}")

t5_tokens = t5_tokenizer.tokenize(complex_word)
print(f"T5 (SentencePiece): {t5_tokens}\n")

# Now let's compare all tokenizers on all test sentences
for sentence in test_sentences:
    compare_subword_tokenizers(sentence)
````

````python
### Simple BPE implementation

def train_bpe(corpus, num_merges):
    """
    Simplified implementation of the BPE algorithm
    """
    # Initialize with character-level vocabulary
    vocab = set()
    # Split each word into characters and add end-of-word symbol
    word_chars = []
    for word in corpus:
        chars = list(word) + ['</w>']  # End-of-word symbol
        word_chars.append(chars)
        vocab.update(chars)
    
    # Keep track of merge operations
    merges = {}
    
    for i in range(num_merges):
        # Count pair frequencies
        pairs = {}
        for word in word_chars:
            for j in range(len(word) - 1):
                pair = (word[j], word[j+1])
                pairs[pair] = pairs.get(pair, 0) + 1
        
        if not pairs:
            break
        
        # Find most frequent pair
        best_pair = max(pairs, key=pairs.get)
        new_token = best_pair[0] + best_pair[1]
        merges[best_pair] = new_token
        vocab.add(new_token)
        
        # Apply the merge to all words
        new_word_chars = []
        for word in word_chars:
            new_word = []
            i = 0
            while i < len(word):
                if i < len(word) - 1 and (word[i], word[i+1]) == best_pair:
                    new_word.append(new_token)
                    i += 2
                else:
                    new_word.append(word[i])
                    i += 1
            new_word_chars.append(new_word)
        word_chars = new_word_chars
        
        print(f"Merge #{i+1}: {best_pair} → {new_token} (frequency: {pairs[best_pair]})")
    
    return vocab, merges, word_chars

# Apply BPE to a small example corpus
corpus = ['low', 'lower', 'lowest', 'newer', 'wider']
vocab, merges, final_words = train_bpe(corpus, 10)

print("\nFinal vocabulary:")
print(sorted(list(vocab)))

print("\nFinal tokenization:")
for word, tokens in zip(corpus, final_words):
    print(f"{word}: {tokens}")
````

### Aside: The Neural Machine Translation Connection

Subword tokenization gained prominence through neural machine translation (NMT). Traditional statistical MT systems struggled with rare words and morphologically rich languages. When a source sentence contained a word not in the translation model's vocabulary, it either had to be left untranslated or replaced with a generic unknown token, severely limiting translation quality.

In 2016, Rico Sennrich and colleagues at the University of Edinburgh adapted Byte Pair Encoding (originally a 1990s data compression algorithm) for NMT in their paper "Neural Machine Translation of Rare Words with Subword Units." The results were dramatic - their BPE approach significantly improved translation of rare words by breaking them into meaningful subwords that could be translated separately and recombined.

For example, the German word "Schwimmbad" (swimming pool) might be broken into "Schwimm" + "bad", where both parts could be translated independently. This worked particularly well for compound words and morphological variants.

The innovation spread quickly as researchers observed that subword tokens often captured linguistically meaningful units like prefixes ("un-", "re-"), stems, and suffixes ("-ing", "-ed"). The timing was perfect - just as NMT was taking off, subword tokenization helped solve one of its biggest limitations.

This synergy between translation needs and tokenization innovation demonstrates how practical problems often drive fundamental algorithmic advances in NLP. Today, virtually all state-of-the-art NLP systems use some form of subword tokenization, and it's become an essential component of language models like BERT, GPT, and T5.

### 2.5 Comparing Tokenization Approaches

Now that we've explored character, word, and subword tokenization, let's compare their properties and understand when to use each approach.

#### Key trade-offs

| Property | Character | Word | Subword |
|----------|-----------|------|--------|
| Vocabulary size | Very small (26-128+) | Very large (10K-1M+) | Moderate (1K-100K) |
| Sequence length | Very long | Short | Moderate |
| OOV handling | No OOV problem | Poor (needs UNK token) | Good (decomposes unknown words) |
| Semantic meaning | Low | High | Medium |
| Language agnostic | Yes | No | Somewhat |
| Computational efficiency | Low | High | Medium |
| Morphological awareness | Emergent | None | Moderate |

#### When to use each approach

- **Character tokenization** is best for:
  - Dealing with noisy text with many spelling errors
  - Languages with no clear word boundaries
  - When vocabulary size must be minimized
  - Character-level tasks like spelling correction

- **Word tokenization** is best for:
  - When semantic units are important
  - Small-to-medium sized corpora with limited vocabulary
  - When training efficiency is crucial
  - Classical NLP approaches

- **Subword tokenization** is best for:
  - Large-scale language models
  - Multilingual applications
  - Handling morphologically rich languages
  - Balancing vocabulary size and semantic meaning

````python
### Comprehensive tokenization comparison

import matplotlib.pyplot as plt
import numpy as np
from collections import Counter

# Load a sample text corpus
nltk.download('gutenberg', quiet=True)
from nltk.corpus import gutenberg

# Use a sample of text from a book
sample_text = ' '.join(gutenberg.sents('austen-emma.txt')[:500])

# Define tokenization functions
def char_tokenize(text):
    return list(text)

def word_tokenize_simple(text):
    return text.split()

def word_tokenize_nltk(text):
    return word_tokenize(text)

def subword_tokenize_bert(text):
    return bert_tokenizer.tokenize(text)

def subword_tokenize_gpt2(text):
    return gpt2_tokenizer.tokenize(text)

# Apply different tokenization methods
tokenizers = {
    "Character": char_tokenize,
    "Simple Word": word_tokenize_simple,
    "NLTK Word": word_tokenize_nltk,
    "BERT (WordPiece)": subword_tokenize_bert,
    "GPT-2 (BPE)": subword_tokenize_gpt2
}

# Collect statistics
stats = {}

for name, tokenize_fn in tokenizers.items():
    tokens = tokenize_fn(sample_text)
    vocab = set(tokens)
    token_counts = Counter(tokens)
    
    stats[name] = {
        "token_count": len(tokens),
        "vocab_size": len(vocab),
        "avg_token_len": np.mean([len(t) for t in tokens]),
        "unique_ratio": len(vocab) / len(tokens),
        "top_tokens": token_counts.most_common(5)
    }

# Display statistics in a table
stats_df = pd.DataFrame({
    "Tokenizer": list(stats.keys()),
    "Sequence Length": [stats[name]["token_count"] for name in stats],
    "Vocabulary Size": [stats[name]["vocab_size"] for name in stats],
    "Avg Token Length": [round(stats[name]["avg_token_len"], 2) for name in stats],
    "Unique Token Ratio": [round(stats[name]["unique_ratio"], 3) for name in stats]
})

print("Tokenization Method Comparison:")
print(stats_df)

# Visualize the trade-offs
fig, ax = plt.subplots(figsize=(12, 6))

tokenizer_names = list(stats.keys())
token_counts = [stats[name]["token_count"] for name in tokenizer_names]
vocab_sizes = [stats[name]["vocab_size"] for name in tokenizer_names]

x = np.arange(len(tokenizer_names))
width = 0.35

# Create logarithmic scale bars for better visualization
ax.bar(x - width/2, token_counts, width, label='Sequence Length')
ax.bar(x + width/2, vocab_sizes, width, label='Vocabulary Size')

ax.set_yscale('log')
ax.set_ylabel('Count (log scale)')
ax.set_title('Tokenization Trade-offs: Sequence Length vs. Vocabulary Size')
ax.set_xticks(x)
ax.set_xticklabels(tokenizer_names)
ax.legend()

plt.tight_layout()
plt.show()

# Show most common tokens for each method
print("\nMost common tokens by method:")
for name in tokenizers:
    print(f"\n{name}: {stats[name]['top_tokens']}")
````

### Aside: Space or No Space

While English uses spaces to separate words, many languages don't follow this convention, creating fundamental challenges for word tokenization.

Chinese, Japanese, and Thai don't use explicit word boundaries in writing. For example, the Chinese sentence "我喜欢自然语言处理" (I like natural language processing) appears as a continuous string of characters with no spaces. This makes word tokenization much more challenging - you can't simply split on whitespace.

Chinese word segmentation uses statistical models and dictionaries to determine word boundaries. For example, the character sequence "自然语言" could be segmented as [自然, 语言] (nature, language) or as a single word [自然语言] (natural language). The correct segmentation depends on context.

Japanese presents an even more complex case with its mixed writing system combining three scripts: kanji (Chinese characters), hiragana, and katakana. Japanese tokenization often relies on specialized tools like MeCab or Juman that combine multiple approaches.

On the other end of the spectrum, German creates compound words without spaces: "Rindfleischetikettierungsüberwachungsaufgabenübertragungsgesetz" is a single word meaning "law for the delegation of monitoring beef labeling." These compounds create vocabulary explosion problems for word tokenizers.

Arabic and Hebrew connect most letters within words and use right-to-left script, presenting unique challenges for tokenization tools designed primarily for left-to-right languages like English.

Korean uses spaces, but between phrases rather than individual words, creating another variation in how "words" are defined.

These differences make universal word tokenization impossible. Language-specific solutions evolved for decades, but the dream of a one-size-fits-all tokenizer led researchers toward more flexible subword approaches, which we can see as a compromise that works reasonably well across many languages.

### 3.1 Principles of Character Tokenization

Character tokenization operates on a simple principle: treat each individual character in the text as a separate token. This approach takes us to the most basic level of written language.

#### Definition and Process

Character tokenization is defined as follows:

Given a text string $S = s_1 s_2 \ldots s_n$, where each $s_i$ is a character, the character tokenization function $f_{char}$ maps $S$ to a sequence of tokens $T = t_1, t_2, \ldots, t_n$ where each token $t_i = s_i$.

In simpler terms, we're just splitting the text into individual characters. This process creates a one-to-one mapping between characters in the original text and tokens in the resulting sequence.

#### Key Properties

1. **Fixed, Finite Vocabulary**: The vocabulary size is bounded by the character set of the language or encoding standard (e.g., ASCII has 128 characters, Unicode has over 143,000).

2. **No Out-of-Vocabulary (OOV) Problem**: Since we're working with individual characters, any properly encoded text can be tokenized without encountering unknown tokens.

3. **Lossless Representation**: Character tokenization preserves all information in the original text, enabling perfect reconstruction.

4. **Long Sequences**: Character sequences are much longer than word sequences for the same text, which can affect computational efficiency.

Let's see a simple example:

````python
def character_tokenize(text):
    """Simple character-level tokenizer.
    
    Args:
        text (str): Input text to tokenize
        
    Returns:
        list: List of character tokens
    """
    return list(text)

# Example usage
sample_text = "Hello, world!"
char_tokens = character_tokenize(sample_text)

print(f"Original text: {sample_text}")
print(f"Character tokens: {char_tokens}")
print(f"Number of tokens: {len(char_tokens)}")
````

### Aside: The Atoms of Language

Characters are often called the "atoms" of written language - the smallest indivisible units that combine to form all text. But this analogy breaks down in interesting ways. Unlike physical atoms, characters can have internal structure (consider letters with diacritics like "é"), and what counts as a "character" varies across writing systems.

In Chinese and Japanese, individual characters (hanzi/kanji) often represent entire concepts or morphemes, making them closer to syllables or even words in information content. One Chinese character might require multiple keystrokes to type and conveys as much information as several Latin characters.

Conversely, in Korean Hangul, what visually appears as a single character is actually a syllable block composed of multiple "jamo" (individual consonants and vowels). The character "한" (han) decomposes into ㅎ+ㅏ+ㄴ.

This variation in what constitutes a "character" has profound implications for NLP across languages. Character tokenization in English processes individual letters, but in Chinese, it's processing entire morphemes. This is one reason why character-level models often perform better on Chinese and Japanese than you might expect - each "character" carries more semantic information.

### 3.2 Unicode and Character Encodings

To properly implement character tokenization, we need to understand how computers represent characters. This brings us to the world of character encodings and Unicode.

#### Character Encoding Evolution

Character encodings map characters to numeric codes that computers can process:

1. **ASCII** (1963): The original 7-bit encoding supporting 128 characters, mostly English letters, numbers, and basic symbols.

2. **ISO-8859 family** (1980s): 8-bit extensions to ASCII supporting various European languages, but incompatible with each other.

3. **Unicode** (1991-present): A universal character encoding standard aiming to support all writing systems. Currently contains over 143,000 characters.

#### Unicode Encoding Formats

Unicode can be implemented through different encoding formats:

- **UTF-8**: Variable-width encoding using 1 to 4 bytes per character. ASCII characters use just 1 byte, making it backwards compatible and space-efficient for English text.

- **UTF-16**: Variable-width encoding using 2 or 4 bytes per character.

- **UTF-32**: Fixed-width encoding using 4 bytes for every character, simplifying character counting but using more storage.

UTF-8 has become the dominant encoding on the web and in modern systems due to its efficiency and compatibility.

#### Unicode Challenges for Tokenization

Unicode introduces several challenges for character tokenization:

1. **Combining Characters**: Some characters can be represented either as a single code point or as a base character followed by combining marks. For example, "é" can be encoded as a single character (U+00E9) or as "e" (U+0065) followed by the combining acute accent (U+0301).

2. **Normalization**: To handle these equivalence issues, Unicode defines normalization forms (NFC, NFD, NFKC, NFKD) that convert between composed and decomposed forms.

3. **Emoji and Zero-Width Joiners**: Modern text often includes emoji, some of which are sequences of multiple code points joined by special invisible characters.

Let's explore how to handle Unicode properly in Python:

````python
import unicodedata

def explore_unicode(text):
    """Explore the Unicode properties of a text string."""
    print(f"Original text: {text}")
    print("Character breakdown:")
    
    for i, char in enumerate(text):
        code_point = f"U+{ord(char):04X}"
        char_name = unicodedata.name(char, "Unknown")
        print(f"  Character {i+1}: '{char}' - {code_point} - {char_name}")
    
    # Check if characters can be decomposed
    nfd_form = unicodedata.normalize('NFD', text)
    if nfd_form != text:
        print("\nNormalized form (NFD - decomposed):")
        for i, char in enumerate(nfd_form):
            code_point = f"U+{ord(char):04X}"
            char_name = unicodedata.name(char, "Unknown")
            print(f"  Character {i+1}: '{char}' - {code_point} - {char_name}")

# Example with accented characters
explore_unicode("café")
print("\n" + "-"*50 + "\n")

# Example with emoji
explore_unicode("Hello 👨‍👩‍👧‍👦!")
````

````python
# A Unicode-aware character tokenizer
def unicode_char_tokenize(text, normalize=True):
    """Unicode-aware character tokenizer.
    
    Args:
        text (str): Input text
        normalize (bool): Whether to normalize the text to NFC form
        
    Returns:
        list: List of character tokens
    """
    if normalize:
        # NFC form composes characters when possible
        text = unicodedata.normalize('NFC', text)
    
    # Grapheme clusters (perceived characters) would be better,
    # but for simplicity we'll use Python's character iteration
    return list(text)

# Example
text_samples = ["café", "café",  # The same word with different encodings
               "北京",           # Chinese
               "こんにちは",      # Japanese
               "안녕하세요"]     # Korean

for text in text_samples:
    tokens = unicode_char_tokenize(text)
    print(f"Text: {text}")
    print(f"Tokens: {tokens}")
    print(f"Token count: {len(tokens)}\n")
````

### 3.3 Advantages and Limitations

Character tokenization has several distinctive advantages and limitations that make it suitable for certain applications while inadequate for others.

#### Advantages of Character Tokenization

1. **No Out-of-Vocabulary (OOV) Problem**
   - Character vocabularies are closed and finite
   - Any properly encoded text can be represented

1. **Long Sequences**

   - Requires more layers or parameters to achieve semantic understanding

3. **Training Inefficiency**
   - More steps to learn meaningful representations
   - Higher computational cost for equivalent text

4. **Unicode Complexity**
   - Combining marks and normalization forms complicate processing

#### Use Cases for Character Tokenization

Character tokenization shines in specific applications:

2. **Spell Checking and Correction**
   - Character-level models can learn common misspellings and suggest corrections

   - Languages where words have complex internal structure

5. **Handling Noisy Text**
   - Social media content, user-generated text
   - OCR output with recognition errors

6. **Languages Without Clear Word Boundaries**
   - Chinese, Japanese, Thai
   - Character tokenization avoids word segmentation problems

````python
# Example: Simple language identification using character n-grams
from collections import Counter
import re

def extract_char_ngrams(text, n=3):
    """Extract character n-grams from text."""
    # Clean and normalize text
    text = unicodedata.normalize('NFC', text.lower())
    text = re.sub(r'\s+', ' ', text)  # Normalize whitespace
    
    # Extract n-grams
    ngrams = [text[i:i+n] for i in range(len(text) - n + 1)]
    return Counter(ngrams)

# Sample texts in different languages
samples = {
    'English': "The quick brown fox jumps over the lazy dog.",
    'Spanish': "El veloz murciélago hindú comía feliz cardillo y kiwi.",
    'French': "Portez ce vieux whisky au juge blond qui fume.",
    'German': "Victor jagt zwölf Boxkämpfer quer über den Sylter Deich."
}

# Extract top n-grams for each language
language_profiles = {}
for lang, text in samples.items():
    ngrams = extract_char_ngrams(text)
    language_profiles[lang] = ngrams.most_common(10)

# Display top n-grams for each language
for lang, ngrams in language_profiles.items():
    print(f"Top 10 character trigrams for {lang}:")
    for ngram, count in ngrams:
        print(f"  '{ngram}': {count}")
    print()

# Simple language identification function
def identify_language(text, language_profiles):
    """Identify the language of a text using character n-gram profiles."""
    text_profile = extract_char_ngrams(text)
    
    # Calculate similarity to each language profile
    similarities = {}
    for lang, profile in language_profiles.items():
        profile_dict = dict(profile)
        similarity = 0
        for ngram, count in text_profile.items():
            if ngram in profile_dict:
                similarity += min(count, profile_dict[ngram])
        similarities[lang] = similarity
    
    # Return the language with the highest similarity
    return max(similarities.items(), key=lambda x: x[1])[0]

# Test the language identifier
test_texts = {
    "English test": "She sells seashells by the seashore.",
    "Spanish test": "Cada estudiante tiene su propio libro.",
    "French test": "Je voudrais un café, s'il vous plaît.",
    "German test": "Ich lerne Deutsch seit drei Jahren."
}

for name, text in test_texts.items():
    detected = identify_language(text, language_profiles)
    print(f"{name}: '{text}'")
    print(f"Detected language: {detected}\n")
````

### Aside: Why Character-Level Models Work

Character-level models have shown surprising effectiveness in many NLP tasks, often performing competitively with word-based approaches despite their apparent simplicity. What explains this success?

The power of character-level models comes from their ability to capture several types of patterns:

First, characters form natural linguistic building blocks. Consider how you process unfamiliar words - you recognize familiar morphemes (prefixes, suffixes, roots) and piece together meaning. When you see "antidisestablishmentarianism," you don't memorize it whole; you break it into "anti-dis-establish-ment-arian-ism" and derive meaning from these parts. Character models can learn similar decompositions.

Second, character patterns often reflect etymology and sound patterns. Romance languages share character patterns, Germanic languages have their own distinctive patterns, and loanwords often retain character sequences from their source language. These patterns provide strong signals for various NLP tasks.

Third, character models excel at handling noise and variation. Humans can easily read "teh" as "the" or understand "l8r" as "later" because we process at multiple levels, including characters. Similarly, character models can learn to be robust to typos, variant spellings, or informal writing.

Andrej Karpathy's influential 2015 blog post "The Unreasonable Effectiveness of Recurrent Neural Networks" demonstrated that character-level models could learn surprisingly sophisticated patterns - from matching brackets and quotes to language-specific conventions - simply from predicting the next character in a sequence. This robustness to surface variation, combined with the ability to learn morphological patterns, makes character-level approaches particularly valuable in an increasingly diverse and informal text landscape.

### 3.4 Character-level Language Models

Character-level language models predict the next character in a sequence given the previous characters. They're a perfect application for character tokenization, and they provide insights into how neural networks can learn patterns from primitive units.

#### Formulation

Given a sequence of characters $c_1, c_2, \ldots, c_t$, a character language model computes the probability of the next character:

$$P(c_{t+1} | c_1, c_2, \ldots, c_t)$$

By the chain rule of probability, we can model the probability of an entire sequence:

$$P(c_1, c_2, \ldots, c_T) = \prod_{t=1}^{T} P(c_t | c_1, c_2, \ldots, c_{t-1})$$

A character language model learns to estimate these conditional probabilities.

#### Architectures for Character-level Models

Several neural architectures work well for character-level modeling:

1. **Recurrent Neural Networks (RNNs)**
   - LSTMs and GRUs process sequences character by character
   - Maintain internal state capturing context from previous characters
   - Popular for character-level generation tasks

2. **Convolutional Neural Networks (CNNs)**
   - Apply convolutions over character sequences
   - Capture local patterns effectively
   - Efficient parallel processing

3. **Transformer-based Models**
   - Self-attention mechanisms capture relationships between characters
   - Effectively model both local and long-range dependencies
   - More compute-intensive but can achieve state-of-the-art results

Let's implement a simple character-level language model using an LSTM architecture:

````python
import torch
import torch.nn as nn
import torch.nn.functional as F
import torch.optim as optim
import numpy as np

class CharLSTM(nn.Module):
    """Character-level language model using LSTM."""
    
    def __init__(self, input_size, hidden_size, output_size, n_layers=1, dropout=0.1):
        """Initialize the model.
        
        Args:
            input_size (int): Size of vocabulary (number of unique characters)
            hidden_size (int): Size of hidden layers
            output_size (int): Size of output layer (same as input_size)
            n_layers (int): Number of LSTM layers
            dropout (float): Dropout probability
        """
        super(CharLSTM, self).__init__()
        
        # Embedding layer converts character indices to dense vectors
        self.embedding = nn.Embedding(input_size, hidden_size)
        
        # LSTM layer
        self.lstm = nn.LSTM(hidden_size, hidden_size, n_layers, 
                           dropout=dropout if n_layers > 1 else 0,
                           batch_first=True)
        
        # Output layer
        self.fc = nn.Linear(hidden_size, output_size)
        
        # Dropout layer
        self.dropout = nn.Dropout(dropout)
        
        self.hidden_size = hidden_size
        self.n_layers = n_layers
    
    def forward(self, x, hidden=None):
        """Forward pass through the network.
        
        Args:
            x: Input tensor of shape (batch_size, seq_length)
            hidden: Initial hidden state
            
        Returns:
            output: Predictions for each character
            hidden: Final hidden state
        """
        # Convert character indices to embeddings
        embedded = self.embedding(x)  # (batch_size, seq_length, hidden_size)
        
        # Apply dropout to embeddings
        embedded = self.dropout(embedded)
        
        # Process sequence through LSTM
        output, hidden = self.lstm(embedded, hidden)  # output: (batch_size, seq_length, hidden_size)
        
        # Apply dropout to LSTM output
        output = self.dropout(output)
        
        # Pass through final fully connected layer
        output = self.fc(output)  # (batch_size, seq_length, output_size)
        
        return output, hidden
    
    def init_hidden(self, batch_size, device):
        """Initialize hidden state."""
        return (torch.zeros(self.n_layers, batch_size, self.hidden_size).to(device),
                torch.zeros(self.n_layers, batch_size, self.hidden_size).to(device))

# Example setup (not enough to actually train the model, but illustrates the architecture)
def char_lm_example():
    # Define parameters
    vocab_size = 128  # ASCII characters
    hidden_size = 256
    n_layers = 2
    
    # Create model
    model = CharLSTM(vocab_size, hidden_size, vocab_size, n_layers)
    
    # Display model architecture
    print(model)
    
    # Example input (batch_size=2, seq_length=10)
    x = torch.randint(0, vocab_size, (2, 10))
    print(f"Input shape: {x.shape}")
    
    # Forward pass
    hidden = model.init_hidden(2, 'cpu')
    output, hidden = model(x, hidden)
    
    print(f"Output shape: {output.shape}")
    print(f"Hidden state shape: {hidden[0].shape}")
    print(f"Cell state shape: {hidden[1].shape}")
    
    # Get predictions
    _, predicted = torch.max(output, dim=2)
    print(f"Predicted shape: {predicted.shape}")

char_lm_example()
````

#### Character-level Generation

Character language models can generate text by sampling the next character repeatedly. Let's implement a simple generation function:

````python
def generate_text(model, seed_text, max_length=500, temperature=1.0, device='cpu'):
    """Generate text using a trained character language model.
    
    Args:
        model: Trained CharLSTM model
        seed_text: Initial text to start generation
        max_length: Maximum number of characters to generate
        temperature: Controls randomness (lower = more deterministic)
        device: Device to run generation on
        
    Returns:
        The generated text
    """
    model.eval()
    generated = seed_text
    hidden = None
    
    # Convert characters to indices (simplified version - assumes ASCII)
    char_to_idx = {chr(i): i for i in range(128)}
    
    # Generate one character at a time
    for i in range(max_length):
        # Convert last character to input tensor
        last_char = generated[-1]
        input_idx = torch.tensor([[char_to_idx.get(last_char, 0)]], device=device)
        
        # Get prediction from model
        with torch.no_grad():
            output, hidden = model(input_idx, hidden)
        
        # Apply temperature to logits
        output = output[:, -1, :] / temperature
        
        # Convert to probabilities
        probs = F.softmax(output, dim=1).squeeze()
        
        # Sample from the distribution
        next_char_idx = torch.multinomial(probs, 1).item()
        next_char = chr(next_char_idx)
        
        # Add to generated text
        generated += next_char
    
    return generated

# This is just a demonstration - we don't have a trained model yet
print("Note: This is a demonstration only. Without a trained model, we can't actually generate text.")
print("In practice, you would train the model on a large corpus before generating text.")
````

#### Training a Character-level Language Model

Training a character language model involves:

1. Preprocessing text into character sequences
2. Converting characters to indices
3. Creating training batches
4. Training the model to predict the next character

Let's outline the training process:

````python
def prep_char_lm_data(text, seq_length=100):
    """Prepare data for character-level language model.
    
    Args:
        text: The input text corpus
        seq_length: Length of sequences to train on
        
    Returns:
        chars: List of all unique characters
        char_to_idx: Dictionary mapping characters to indices
        idx_to_char: Dictionary mapping indices to characters
        x_data: Input sequences
        y_data: Target sequences (next character for each position)
    """
    # Get unique characters
    chars = sorted(list(set(text)))
    
    # Create mapping dictionaries
    char_to_idx = {ch: i for i, ch in enumerate(chars)}
    idx_to_char = {i: ch for i, ch in enumerate(chars)}
    
    # Convert text to indices
    indices = [char_to_idx[ch] for ch in text]
    
    # Create sequences
    x_data = []
    y_data = []
    
    for i in range(0, len(indices) - seq_length):
        x_data.append(indices[i:i + seq_length])
        y_data.append(indices[i + 1:i + seq_length + 1])  # Shifted by 1 for next-character prediction
    
    # Convert to PyTorch tensors
    x_data = torch.tensor(x_data)
    y_data = torch.tensor(y_data)
    
    return chars, char_to_idx, idx_to_char, x_data, y_data

def train_char_lm(model, x_data, y_data, n_epochs=10, batch_size=128, lr=0.001, device='cpu'):
    """Train a character language model.
    
    Args:
        model: CharLSTM model
        x_data: Input sequences
        y_data: Target sequences
        n_epochs: Number of training epochs
        batch_size: Batch size
        lr: Learning rate
        device: Device to train on
        
    Returns:
        List of training losses
    """
    model.to(device)
    optimizer = optim.Adam(model.parameters(), lr=lr)
    criterion = nn.CrossEntropyLoss()
    
    # Prepare data
    dataset = torch.utils.data.TensorDataset(x_data, y_data)
    dataloader = torch.utils.data.DataLoader(dataset, batch_size=batch_size, shuffle=True)
    
    losses = []
    
    for epoch in range(n_epochs):
        model.train()
        epoch_loss = 0
        
        for x_batch, y_batch in dataloader:
            # Move tensors to device
            x_batch = x_batch.to(device)
            y_batch = y_batch.to(device)
            
            # Initialize hidden state
            hidden = model.init_hidden(x_batch.size(0), device)
            
            # Zero gradients
            optimizer.zero_grad()
            
            # Forward pass
            output, _ = model(x_batch, hidden)
            
            # Reshape for loss calculation
            output = output.view(-1, output.size(2))  # (batch_size * seq_len, vocab_size)
            y_batch = y_batch.view(-1)  # (batch_size * seq_len)
            
            # Calculate loss
            loss = criterion(output, y_batch)
            
            # Backward pass and optimization
            loss.backward()
            # Clip gradients to prevent exploding gradients
            torch.nn.utils.clip_grad_norm_(model.parameters(), 5)
            optimizer.step()
            
            epoch_loss += loss.item()
        
        # Average loss for the epoch
        avg_loss = epoch_loss / len(dataloader)
        losses.append(avg_loss)
        
        print(f"Epoch {epoch+1}/{n_epochs}, Loss: {avg_loss:.4f}")
    
    return losses

# This is just a demonstration - we don't have a text corpus to train on here
print("Note: This is a training process outline only. In practice, you would need a large text corpus.")
````

### 4.2 Rule-based Word Tokenizers

Rule-based tokenizers rely on explicit rules crafted by humans to define word boundaries. These rules typically involve:

1. **Whitespace splitting**: The most basic approach
2. **Punctuation handling**: Rules for separating or keeping punctuation
3. **Special case handling**: Rules for contractions, hyphenation, abbreviations, etc.

The Penn Treebank tokenization standard is one of the most influential rule sets in English NLP. It specifies handling for contractions ("don't" → "do n't"), punctuation (separate most punctuation marks), and other special cases.

Rule-based tokenizers are straightforward to implement and debug, but they're language-specific and may require constant maintenance to handle new cases.

````python
# Implementing a simple rule-based tokenizer
import re

def simple_rule_tokenizer(text):
    """A simple rule-based tokenizer with some basic rules for English."""
    # Step 1: Replace contractions with separate tokens
    text = re.sub(r"([a-zA-Z])'([a-z])", r"\1 '\2", text)  # don't -> don 't
    
    # Step 2: Separate punctuation from words
    text = re.sub(r"([.,!?;():"\[\]{}])", r" \1 ", text)
    
    # Step 3: Handle hyphenated words (keep together)
    # This is simplified - a real implementation would be more complex
    
    # Step 4: Split on whitespace and filter empty strings
    tokens = [token for token in text.split() if token]
    
    return tokens

# Test our tokenizer
sample_texts = [
    "I don't like that, it's too expensive!",
    "The state-of-the-art model achieved 97.5% accuracy.",
    "Email me at user@example.com or visit https://example.com."
]

for text in sample_texts:
    print(f"Original: {text}")
    print(f"Tokenized: {simple_rule_tokenizer(text)}")
    print()
````

### 4.3 Statistical Word Tokenizers

For languages without clear word boundaries (like Chinese, Japanese, or Thai), rule-based approaches are insufficient. Statistical tokenizers use probability models to predict word boundaries. These methods:

1. **Learn from data**: They're trained on pre-tokenized text
2. **Use context**: They consider surrounding characters when making decisions
3. **Apply probability**: They choose the most likely segmentation

Common approaches include:

- **Maximum Matching (MaxMatch)**: Greedy algorithm that tries to match the longest dictionary word first
- **Hidden Markov Models (HMMs)**: Model state transitions between word/non-word boundaries
- **Conditional Random Fields (CRFs)**: Model the conditional probability of boundary decisions
- **Neural approaches**: Use recurrent or convolutional networks to predict boundaries

Let's look at a simple example of MaxMatch for Chinese tokenization:

````python
# Simple MaxMatch algorithm for Chinese word segmentation
def max_match(text, dictionary, max_word_length=5):
    """Implements a simple Maximum Matching algorithm.
    Args:
        text: The text to segment
        dictionary: A set of known words
        max_word_length: Maximum word length to consider
    Returns:
        A list of segmented words
    """
    words = []
    while text:
        # Try to match the longest possible word
        word = None
        for i in range(min(max_word_length, len(text)), 0, -1):
            if text[:i] in dictionary:
                word = text[:i]
                break
        
        # If no match found, take the first character as a word
        if word is None:
            word = text[0]
            
        words.append(word)
        text = text[len(word):]
    
    return words

# Simple Chinese dictionary (very limited for demonstration)
chinese_dict = {'我', '是', '学生', '学', '生', '喜欢', '喜', '欢', '自然', '语言', '处理', '自', '然', '语', '言', '处', '理'}

# Example Chinese text: "我是学生喜欢自然语言处理" (I am a student who likes natural language processing)
text = "我是学生喜欢自然语言处理"

# Apply MaxMatch
segmented = max_match(text, chinese_dict)
print(f"Original: {text}")
print(f"Segmented: {' '.join(segmented)}")

# Using a real Chinese tokenizer (jieba)
import jieba
print(f"\njieba segmentation: {' '.join(jieba.cut(text))}")
````

### 4.5 Stemming and Lemmatization

Once we've tokenized text into words, we often want to normalize different forms of the same word. For example, "running", "runs", and "ran" all derive from "run". Two main approaches address this:

#### Stemming

Stemming uses rule-based algorithms to chop off word endings, producing a stem that may not be a valid word:

- **Porter Stemmer**: Classic algorithm with 5 phases of rewrite rules
- **Snowball (Porter2)**: Improved version of Porter
- **Lancaster**: More aggressive stemming algorithm

Examples:
- "running", "runner", "runs" → "run"
- "argued", "argues", "arguing" → "argu" (not a valid word)

#### Lemmatization

Lemmatization uses linguistic knowledge to convert words to their dictionary form (lemma), always producing valid words:

- Requires knowledge of parts of speech
- Uses dictionary lookups and morphological analysis
- More accurate but slower than stemming

Examples:
- "running", "runs" → "run"
- "better" → "good"
- "argued", "argues" → "argue"

Let's compare stemming and lemmatization:

````python
from nltk.stem import PorterStemmer, SnowballStemmer, WordNetLemmatizer
import nltk

# Download necessary NLTK data
nltk.download('wordnet', quiet=True)
nltk.download('omw-1.4', quiet=True)

# Initialize stemmers and lemmatizer
porter = PorterStemmer()
snowball = SnowballStemmer('english')
lemmatizer = WordNetLemmatizer()

# Sample words to compare
test_words = [
    "running", "runs", "ran",
    "better", "best", "good",
    "argued", "argues", "arguing",
    "mice", "mouse",
    "flying", "flies", "flew"
]

# Compare the results
results = []
for word in test_words:
    results.append({
        "Word": word,
        "Porter Stem": porter.stem(word),
        "Snowball Stem": snowball.stem(word),
        "Lemma (noun)": lemmatizer.lemmatize(word, 'n'),  # As noun
        "Lemma (verb)": lemmatizer.lemmatize(word, 'v')   # As verb
    })

pd.DataFrame(results)
````

Notice how:
- Stemming is more aggressive and sometimes produces non-words
- Lemmatization produces valid words but requires part-of-speech information
- Irregular forms ("mice" → "mouse") are handled correctly by lemmatization but not by stemming

Now let's see how these affect a real text analysis task:

````python
# Compare raw, stemmed, and lemmatized text in a frequency analysis
from collections import Counter

text = """
The runner runs every morning. He ran yesterday too.
Running is good for health, but better not overdo it.
The best runners know when to rest. They argue that
resting is as important as running. I argued with my
trainer about this, but he was arguing from experience.
"""

# Tokenize
tokens = nltk.word_tokenize(text.lower())
tokens = [t for t in tokens if t.isalpha()]  # Keep only alphabetic tokens

# Process with each method
raw_tokens = tokens
stemmed_porter = [porter.stem(t) for t in tokens]
stemmed_snowball = [snowball.stem(t) for t in tokens]

# Lemmatization with proper POS tagging
nltk.download('averaged_perceptron_tagger', quiet=True)
pos_tags = nltk.pos_tag(tokens)

# Convert NLTK POS tags to WordNet POS tags
def get_wordnet_pos(tag):
    if tag.startswith('J'):
        return 'a'  # Adjective
    elif tag.startswith('V'):
        return 'v'  # Verb
    elif tag.startswith('N'):
        return 'n'  # Noun
    elif tag.startswith('R'):
        return 'r'  # Adverb
    else:
        return 'n'  # Default to noun

lemmatized = [lemmatizer.lemmatize(word, get_wordnet_pos(tag)) 
              for word, tag in pos_tags]

# Count frequencies
raw_freq = Counter(raw_tokens)
stemmed_porter_freq = Counter(stemmed_porter)
stemmed_snowball_freq = Counter(stemmed_snowball)
lemmatized_freq = Counter(lemmatized)

# Display results
print("Raw tokens frequency:")
print(dict(raw_freq.most_common(10)))
print("\nPorter stemming frequency:")
print(dict(stemmed_porter_freq.most_common(10)))
print("\nSnowball stemming frequency:")
print(dict(stemmed_snowball_freq.most_common(10)))
print("\nLemmatization frequency:")
print(dict(lemmatized_freq.most_common(10)))
````

### Aside: The Penn Treebank Standard

Much of English NLP tokenization follows conventions established by the Penn Treebank in the 1990s. This standard includes specific rules like separating contractions (don't → do n't), keeping hyphenated words together, and splitting punctuation from words. ### Aside: The Vocabulary Explosion

Word-based models suffer from vocabulary explosion. English has ~170,000 words in current use, but with inflections, proper names, and domain terminology, NLP systems might encounter millions of unique word forms. Each requires its own embedding vector, consuming memory and computation. Rare words receive poor representations due to limited examples.

Technical domains exacerbate this problem:
- **Biomedical texts** contain specialized terminology
- **Social media** has creative spellings and neologisms
- **Legal documents** use archaic terms

Word-level models typically limit vocabulary to 50,000-100,000 most frequent words, discarding or replacing rarer terms - an unsatisfying compromise that motivated subword approaches.

This vocabulary explosion creates the long-tail problem, where a large portion of any corpus consists of words that appear very rarely. This follows Zipf's Law: the frequency of a word is inversely proportional to its rank in the frequency table.

![Zipf's Law Word Frequency](https://miro.medium.com/max/1400/1*l41IMtkiX56P780R9R1EIw.png)
*Zipf's Law showing word frequency distribution on a log-log scale, highlighting the long tail problem in vocabulary*

````python
# Demonstrating Zipf's Law in a corpus
import matplotlib.pyplot as plt
from nltk.corpus import brown
import numpy as np
nltk.download('brown', quiet=True)

# Get all words from the Brown corpus (lowercased)
words = [word.lower() for word in brown.words()]

# Count word frequencies
word_counts = Counter(words)

# Get frequencies and ranks
freqs = [count for _, count in word_counts.most_common()]
ranks = range(1, len(freqs) + 1)

# Plot on a log-log scale
plt.figure(figsize=(10, 6))
plt.loglog(ranks, freqs, marker='.', linestyle='none', alpha=0.5)

# Add a best-fit line for Zipf's Law
log_ranks = np.log(ranks)
log_freqs = np.log(freqs)
coeffs = np.polyfit(log_ranks, log_freqs, 1)
fit_line = np.exp(coeffs[1]) * np.array(ranks) ** coeffs[0]
plt.loglog(ranks, fit_line, 'r-', label=f'Slope: {coeffs[0]:.2f}')

plt.title("Zipf's Law in the Brown Corpus")
plt.xlabel('Word Rank')
plt.ylabel('Frequency')
plt.legend()
plt.grid(True, alpha=0.3)

# Annotate some interesting words
top_words = [word for word, _ in word_counts.most_common(5)]
rare_words = [word for word, count in word_counts.items() if count == 1][:5]  # Some hapax legomena
middle_words = [word for word, count in word_counts.items() if count == 10][:5]  # Some medium-frequency words

# Select some words to annotate
words_to_annotate = top_words + middle_words + rare_words
for word in words_to_annotate:
    rank = list(word_counts.keys()).index(word) + 1
    freq = word_counts[word]
    plt.annotate(word, (rank, freq), alpha=0.7, fontsize=8)

plt.show()

# Print some statistics
print(f"Total tokens in corpus: {len(words)}")
print(f"Unique words (vocabulary size): {len(word_counts)}")
print(f"Number of words appearing only once: {sum(1 for _, count in word_counts.items() if count == 1)}")
print(f"Percentage of vocabulary appearing only once: {sum(1 for _, count in word_counts.items() if count == 1) / len(word_counts) * 100:.2f}%")
````

This demonstration illustrates the vocabulary explosion problem - a large percentage of the vocabulary appears only once or a few times in the corpus. This makes it hard for word-based models to learn good representations for rare words.

## Section 5: Subword Tokenization

We've explored character tokenization and word tokenization, each with their own strengths and limitations. Character tokenization gives us complete coverage but creates very long sequences, while word tokenization creates more manageable sequences but struggles with out-of-vocabulary (OOV) words. Is there a middle ground?

**Enter subword tokenization** - one of the most important innovations in modern NLP. Subword tokenization breaks words into meaningful units larger than characters but smaller than entire words. This approach gives us the best of both worlds: the vocabulary size efficiency of word tokenization with the coverage of character tokenization.

In this section, we'll explore how subword tokenization works, implement key algorithms like Byte Pair Encoding, and understand why almost all modern language models (BERT, GPT, T5, etc.) use subword tokenization as their foundation.

### The Intuition Behind Subword Tokenization

Consider the word "unhappiness":

- As a single word token: "unhappiness" (requires this exact word in vocabulary)
- As character tokens: ["u", "n", "h", "a", "p", "p", "i", "n", "e", "s", "s"] (11 tokens)
- As subword tokens: ["un", "happiness"] or perhaps ["un", "happy", "ness"] (2-3 tokens)

Languages naturally contain meaningful subword units. In English, we have:
- Prefixes: "un-", "re-", "pre-", etc.
- Stems: "happy", "help", "play", etc.
- Suffixes: "-ness", "-ing", "-ed", etc.

By tokenizing at this level, we:
1. Keep the vocabulary manageable (typically 10k-50k tokens vs. 200k+ for words)
2. Handle any word, even previously unseen ones
3. Maintain reasonable sequence lengths
4. Capture meaningful morphological units

### 5.1 The Need for Subword Tokenization

Why did NLP researchers develop subword tokenization? Let's consider the limitations of previous approaches:

#### The Problem with Word Tokenization

1. **Vocabulary Explosion**: Languages can have hundreds of thousands of unique words, requiring massive embedding matrices.
   
2. **Out-Of-Vocabulary (OOV) Problem**: Any word not seen during training gets mapped to a generic `<UNK>` (unknown) token, losing critical information.
   
3. **Morphologically Rich Languages**: Languages like Finnish, Turkish, or German can create extremely long compound words that are rarely repeated exactly.

#### The Problem with Character Tokenization

1. **Long Sequences**: Character sequences are 5-10x longer than word sequences, increasing computational burden.
   
2. **Lost Efficiency**: Characters themselves often lack semantic meaning, requiring the model to rediscover word structure.

#### Example: Medical Text

Consider medical terminology:

The word "**hyperglycemia**" might be unseen in training, but with subword tokenization, we can break it into ["hyper", "glyc", "emia"], connecting it to other medical terms like "**hyperthermia**" ["hyper", "therm", "ia"] and "**glycemic**" ["glyc", "emic"].

Subwords allow models to handle specialized vocabulary by recognizing meaningful components, even when whole words are unseen. This is particularly important in domains like medicine, law, and science with extensive specialized terminology.

#### The Sweet Spot

Subword tokenization finds the optimal balance by:
1. Learning frequent character sequences as single tokens
2. Falling back to smaller units for rare patterns
3. Adapting automatically to the training data statistics

### 5.2 Byte Pair Encoding (BPE) Algorithm

Byte Pair Encoding (BPE) is one of the most widely used subword tokenization algorithms. Originally developed for data compression in the 1990s, it was adapted for NLP by Sennrich et al. in 2016 for Neural Machine Translation.

#### How BPE Works

The algorithm has two main phases:

1. **Training Phase** - Learn a vocabulary of subwords:
   - Start with a vocabulary of individual characters
   - Count all adjacent character pairs in your training corpus
   - Merge the most frequent pair, adding the new merged unit to the vocabulary
   - Repeat until desired vocabulary size is reached

2. **Tokenization Phase** - Apply the learned vocabulary:
   - Start with a sequence of characters
   - Apply the learned merges in the same order they were learned until no more merges are possible

#### BPE Training Example

Let's walk through a simplified example of training BPE on a tiny corpus:

```
Corpus: "low lower lowest lowest lowest"
```

Initial character vocabulary: {"l", "o", "w", "e", "r", " ", "s", "t"}

1. Count pairs: (l,o):5, (o,w):5, (w,e):3, (e,r):3, etc.
2. Most frequent pair: (l,o) → merge to "lo"
3. Update corpus: "lo w lo wer lo west lo west lo west"
4. Count pairs: (lo,w):5, (w,e):3, etc.
5. Most frequent pair: (lo,w) → merge to "low"
6. Update corpus: "low low er low est low est low est"

And so on until we reach our target vocabulary size.

After training, tokenizing a new word like "slowest" would involve checking which merges can be applied in the order they were learned, perhaps resulting in tokens like ["s", "low", "est"].

#### Why BPE Works Well

BPE adapts well to the data because:

1. Frequent words remain intact as single tokens
2. Rare words get broken down into meaningful subwords
3. The vocabulary automatically discovers morphological patterns

This approach scales effectively across languages and domains, making it a cornerstone of modern NLP.

````python
### BPE Implementation

import re
from collections import defaultdict, Counter

class SimpleBPE:
    def __init__(self, vocab_size):
        self.vocab_size = vocab_size
        self.vocab = {}
        self.merges = {}
        self.split_pattern = None
    
    def train(self, texts):
        # Initialize with character vocabulary
        word_counts = Counter()
        for text in texts:
            words = text.split()
            word_counts.update(words)
        
        # Split words into characters
        splits = {}
        for word, count in word_counts.items():
            chars = list(word)
            splits[word] = chars
        
        # Get initial vocabulary (unique characters)
        vocab = set()
        for word in splits.values():
            for char in word:
                vocab.add(char)
        
        # Initialize merges with character pairs in the corpus
        merges = {}
        
        # Add character vocabulary to final vocab
        self.vocab = {c: i for i, c in enumerate(vocab)}
        
        # BPE training loop
        while len(self.vocab) < self.vocab_size:
            # Count pairs
            pair_counts = defaultdict(int)
            for word, count in word_counts.items():
                chars = splits[word]
                for i in range(len(chars) - 1):
                    pair = (chars[i], chars[i + 1])
                    pair_counts[pair] += count
            
            if not pair_counts:
                break
                
            # Find most frequent pair
            best_pair = max(pair_counts.items(), key=lambda x: x[1])[0]
            new_token = best_pair[0] + best_pair[1]
            
            # Add to vocabulary and merges
            self.vocab[new_token] = len(self.vocab)
            self.merges[best_pair] = new_token
            
            # Apply the merge to all words
            for word in splits:
                chars = splits[word]
                i = 0
                while i < len(chars) - 1:
                    if i < len(chars) - 1 and (chars[i], chars[i+1]) == best_pair:
                        chars[i] = new_token
                        del chars[i+1]
                    else:
                        i += 1
                splits[word] = chars
        
        # Create regex pattern for tokenization
        vocab_tokens = list(self.vocab.keys())
        vocab_tokens.sort(key=len, reverse=True)
        self.split_pattern = re.compile('|'.join(re.escape(t) for t in vocab_tokens))
        
        return self.vocab, self.merges
    
    def tokenize(self, text):
        return self.split_pattern.findall(text)

# Example usage
corpus = [
    "low lower lowest",
    "slow slower slowest", 
    "happy happier happiest"
]

bpe = SimpleBPE(vocab_size=30)
bpe.train(corpus)

print("Vocabulary:")
print(sorted(bpe.vocab.items(), key=lambda x: x[1]))

print("\nExample tokenization:")
print(bpe.tokenize("happiest"))
print(bpe.tokenize("lowest"))
print(bpe.tokenize("unknown")) # Test on unseen word
````

### Aside: The Compression Connection

BPE has a fascinating history that many NLP practitioners aren't aware of. It wasn't developed for language processing at all!

Byte Pair Encoding was originally created by Philip Gage in 1994 as a data compression algorithm. In its original form, BPE works by replacing the most frequent pair of consecutive bytes in the data with a single, unused byte. This process continues until no more replacements are possible, effectively compressing the data.

The NLP adaptation of BPE, introduced by Sennrich et al. in 2016, maintains the core merging strategy but with a crucial difference: instead of creating new bytes, it keeps the merged units readable as subwords. This compression heritage explains both BPE's strengths and limitations:

- Like good compression algorithms, it efficiently identifies and represents common patterns in the data
- It adapts to the statistical properties of any corpus, language-agnostic
- However, it lacks linguistic awareness and sometimes creates subwords that cross meaningful boundaries

The connection to compression algorithms continues with modern models. GPT-3 uses a variant called 'Byte-level BPE' that operates on raw bytes instead of unicode characters, allowing it to encode literally any string without special handling for different languages or character sets.

This cross-pollination of ideas between data compression and NLP demonstrates how breakthroughs often come from applying techniques from one field to problems in another. The best solutions aren't always developed specifically for the problem at hand!

### 5.3 WordPiece and SentencePiece

While BPE is foundational, other important subword tokenization algorithms have been developed with slightly different approaches and trade-offs.

#### WordPiece

WordPiece was developed by Google and rose to prominence through its use in BERT. Though similar to BPE, it has a key difference:

- BPE selects merges based on **frequency** (most frequent pair wins)
- WordPiece selects merges based on **likelihood** (chooses the pair that maximizes the probability of the training data when joined)

This likelihood-based approach tends to favor merges that create meaningful subword units.

WordPiece also uses a different notation convention than BPE:
- Words are split into wordpieces
- Wordpieces that don't begin words start with '##'

For example: "playing" → ["play", "##ing"]

This marking makes it easy to reconstruct the original text by removing the '##' prefixes and concatenating.

#### SentencePiece

SentencePiece, developed by Google, addresses a fundamental limitation of most tokenizers: they rely on pre-tokenization (usually whitespace-based) to identify word boundaries. This approach fails for languages without explicit word separators like Chinese and Japanese.

Key innovations of SentencePiece:

1. **Direct raw text processing**: Treats input as a raw Unicode sequence with no pre-tokenization
2. **Space as a normal symbol**: Encodes spaces just like any other character
3. **Multiple algorithm support**: Implements both BPE and Unigram Language Model tokenization
4. **Reversibility**: Guarantees perfect round-trip conversion from text to tokens and back

SentencePiece has become particularly important for multilingual models that must handle diverse writing systems equally well.

Here's how different tokenizers would process the same sentence:

```
Text: "I love natural language processing!"

BPE: ["I", "love", "natural", "language", "process", "ing", "!"]
WordPiece: ["I", "love", "natural", "language", "process", "##ing", "!"]
SentencePiece: ["▁I", "▁love", "▁natural", "▁language", "▁processing", "!"]
```

Notice how SentencePiece uses the special character "▁" (U+2581) to mark the beginning of words, rather than marking continuations. This approach works for any language, regardless of its writing system.

### 5.4 Unigram Language Model Tokenization

The Unigram Language Model approach to subword tokenization represents another important alternative to BPE and WordPiece. It was developed as part of SentencePiece but has become an algorithm in its own right.

#### How Unigram Works

Unlike BPE and WordPiece which start with character tokens and build up, Unigram starts with a large vocabulary and iteratively prunes it down:

1. **Initialize**: Start with a reasonably large vocabulary (e.g., all possible substrings in the corpus up to some length)

2. **Define model**: A unigram language model where the probability of a token sequence is the product of the individual token probabilities

3. **Train loop**:
   - Calculate loss over the corpus using current vocabulary
   - Compute how much loss would increase if each token were removed
   - Remove the tokens that cause the least increase in loss (typically bottom 10-20%)
   - Repeat until target vocabulary size is reached

4. **Tokenization**: For new text, find the most probable segmentation under the learned model

#### Probabilistic Foundation

The key difference is that Unigram uses a probabilistic model. For a given word, there may be multiple valid tokenizations, and Unigram assigns probabilities to each:

```
Word: "playing"
Possible tokenizations: 
["play", "ing"] - probability 0.8
["playing"] - probability 0.1
["p", "laying"] - probability 0.05
["p", "lay", "ing"] - probability 0.05
```

During inference, Unigram can either:
- Choose the most likely tokenization (Viterbi algorithm)
- Sample from possible tokenizations according to their probabilities

This probabilistic foundation makes Unigram more flexible but also more computationally intensive than the greedy approaches used by BPE and WordPiece.

### 5.5 Subword Tokenization in Modern Language Models

Subword tokenization is foundational to virtually all modern language models. Let's examine how different models implement it:

#### BERT

BERT uses a WordPiece tokenizer with several distinctive features:

- Vocabulary size: 30,522 tokens
- Special tokens: `[CLS]`, `[SEP]`, `[MASK]`, `[UNK]`
- Continuation marker: `##` prefix
- Example: "playing" → ["play", "##ing"]

#### GPT-2 and GPT-3

The GPT family uses a specialized variant called Byte-Level BPE:

- Works on raw bytes rather than unicode characters
- Can encode *any* string without special handling
- Vocabulary includes characters corresponding to every possible byte value (0-255)
- GPT-2 vocabulary size: 50,257 tokens
- GPT-3 vocabulary size: 50,257 tokens (same tokenizer as GPT-2)

#### T5

T5 uses the SentencePiece tokenizer with Unigram algorithm:

- Vocabulary size: 32,000 tokens
- No explicit word segmentation in preprocessing
- Uses the "▁" character to mark word beginnings
- Selected for its language-agnostic properties (important for T5's multilingual capabilities)

#### RoBERTa

RoBERTa uses a byte-level BPE similar to GPT-2:

- Vocabulary size: 50,265 tokens
- Does not use the `##` continuation marker like BERT
- Optimized for efficiency and consistency

#### XLM-R

The multilingual XLM-RoBERTa model pushed vocabulary size much larger:

- Vocabulary size: 250,002 tokens
- SentencePiece with BPE algorithm
- Large vocabulary needed to cover 100 languages effectively

#### Impact on Model Performance

The choice of tokenizer and vocabulary size significantly impacts model performance:

1. **Vocabulary size trade-offs**: Larger vocabularies reduce sequence lengths and can better handle rare words, but increase embedding matrix size and may lead to data sparsity.

2. **Language coverage**: Different tokenization approaches handle multilingual text with varying degrees of efficiency.

3. **Special domain text**: Scientific, legal, or social media text may benefit from domain-specific tokenization.

Researchers have found that simply changing tokenization parameters while keeping the model architecture identical can lead to substantial performance differences on downstream tasks.

This is a classic example of how engineering decisions that seem minor at first can shape a field's development. Many researchers who use BERT's tokenization don't fully understand the algorithm behind it, yet it's become part of the standard toolkit through the model's success and influence.

### 6.1 From Symbolic to Distributed Representations

At the heart of natural language processing lies a fundamental question: how do we represent words computationally? The journey from symbolic to distributed representations marks one of the most significant paradigm shifts in the field.

#### Symbolic Representations

Early NLP systems used **symbolic representations**—discrete symbols with no inherent meaning. In this approach, words are atomic units with no relationship to other words except what is explicitly encoded in rules or taxonomies.

```
word = symbol
king = king
queen = queen
```

The computer sees these as completely unrelated symbols, with no sense that "king" and "queen" might be related.

#### The Limitation of Symbols

The symbolic approach faces a critical problem: **similarity is not captured**. There's no natural way to express that "dog" is more similar to "puppy" than to "laptop." This made language understanding brittle and unable to generalize to new examples.

#### Distributed Representations

In contrast, **distributed representations** encode meaning across multiple dimensions. Each word is represented as a vector in a high-dimensional space, where:

- Each dimension potentially captures a semantic or syntactic feature
- Meaning is distributed across all dimensions
- Similar words have similar vectors

This approach allows us to mathematically capture relationships between words and perform operations on meaning itself!

### 6.2 One-Hot Encoding vs. Dense Embeddings

Let's compare the two main approaches to representing words numerically: one-hot encodings and dense embeddings.

#### One-Hot Encoding

One-hot encoding represents each word as a vector with a single 1 and all other elements 0. The vector length equals the vocabulary size.

For a vocabulary {"cat", "dog", "mouse"}, the one-hot encodings would be:
- "cat" = [1, 0, 0]
- "dog" = [0, 1, 0]
- "mouse" = [0, 0, 1]

**Limitations of One-Hot Encoding:**
1. **Dimensionality**: For real vocabularies (50,000+ words), these become enormous, sparse vectors
2. **No semantics**: All words are equally distant from each other (orthogonal vectors)
3. **No generalization**: No way to infer relationships between words

#### Dense Embeddings

Dense embeddings represent words as compact, dense vectors (typically 50-300 dimensions), learned from data:

- "cat" = [0.2, -0.4, 0.7, ...]
- "dog" = [0.1, -0.3, 0.8, ...]
- "mouse" = [0.4, -0.5, 0.3, ...]

**Advantages of Dense Embeddings:**
1. **Compact**: Much lower dimensionality (e.g., 300 vs. 50,000+)
2. **Semantic**: Similar words have similar vectors
3. **Generalizable**: Relationships between words are captured geometrically

![One-hot vs Dense Embeddings](https://i.imgur.com/NLKJCVr.jpg)
*Comparison between high-dimensional, sparse one-hot encodings and compact, dense word embeddings. Note how in dense representations, semantically similar words cluster together.*

````python
### Code: Implementing One-Hot Encoding vs. Using Dense Embeddings

import numpy as np
import matplotlib.pyplot as plt
from sklearn.manifold import TSNE

# Simple vocabulary
vocabulary = ["king", "queen", "man", "woman", "child", "royal", "palace"]
vocab_size = len(vocabulary)
word_to_idx = {word: idx for idx, word in enumerate(vocabulary)}

# One-hot encoding implementation
def one_hot_encode(word, word_to_idx, vocab_size):
    """Convert a word to a one-hot encoded vector"""
    vector = np.zeros(vocab_size)
    vector[word_to_idx[word]] = 1
    return vector

# Create one-hot encodings
one_hot_vectors = {word: one_hot_encode(word, word_to_idx, vocab_size) 
                  for word in vocabulary}

# Display a few examples
print("One-hot encoding examples:")
for word in ["king", "queen", "man"]:
    print(f"{word}: {one_hot_vectors[word]}")

# Calculate similarities between one-hot vectors
def cosine_similarity(v1, v2):
    """Compute cosine similarity between two vectors"""
    dot_product = np.dot(v1, v2)
    norm_v1 = np.linalg.norm(v1)
    norm_v2 = np.linalg.norm(v2)
    return dot_product / (norm_v1 * norm_v2) if norm_v1 * norm_v2 != 0 else 0

print("\nOne-hot cosine similarities:")
print(f"king-queen: {cosine_similarity(one_hot_vectors['king'], one_hot_vectors['queen'])}")
print(f"king-man: {cosine_similarity(one_hot_vectors['king'], one_hot_vectors['man'])}")
print(f"queen-woman: {cosine_similarity(one_hot_vectors['queen'], one_hot_vectors['woman'])}")

# Simulate dense embeddings (in practice, these would be learned from data)
np.random.seed(42)
embedding_dim = 5  # Much smaller than vocabulary size

# Create related embeddings for demonstration
base_vectors = {
    "royal": np.random.randn(embedding_dim),
    "male": np.random.randn(embedding_dim),
    "female": np.random.randn(embedding_dim)
}

# Create embeddings with deliberate relationships
dense_vectors = {
    "king": base_vectors["royal"] + base_vectors["male"],
    "queen": base_vectors["royal"] + base_vectors["female"],
    "man": base_vectors["male"] + np.random.randn(embedding_dim) * 0.1,
    "woman": base_vectors["female"] + np.random.randn(embedding_dim) * 0.1,
    "child": np.random.randn(embedding_dim),
    "royal": base_vectors["royal"] + np.random.randn(embedding_dim) * 0.1,
    "palace": base_vectors["royal"] + np.random.randn(embedding_dim) * 0.3
}

# Show a few examples
print("\nDense embedding examples (5 dimensions):")
for word in ["king", "queen", "man"]:
    print(f"{word}: {dense_vectors[word]}")

# Calculate similarities between dense vectors
print("\nDense vector cosine similarities:")
print(f"king-queen: {cosine_similarity(dense_vectors['king'], dense_vectors['queen'])}")
print(f"king-man: {cosine_similarity(dense_vectors['king'], dense_vectors['man'])}")
print(f"queen-woman: {cosine_similarity(dense_vectors['queen'], dense_vectors['woman'])}")

# Demonstrate vector arithmetic (king - man + woman ≈ queen)
arithmetic_result = dense_vectors['king'] - dense_vectors['man'] + dense_vectors['woman']
similarities = {word: cosine_similarity(arithmetic_result, vec) 
               for word, vec in dense_vectors.items()}

print("\nVector arithmetic: king - man + woman ≈ ?")
sorted_words = sorted(similarities.items(), key=lambda x: x[1], reverse=True)
for word, similarity in sorted_words:
    print(f"{word}: {similarity:.4f}")
````

When word embeddings emerged, their ability to perform 'vector arithmetic' seemed almost magical. The classic example: vector('king') - vector('man') + vector('woman') ≈ vector('queen'). Similar patterns emerged for countries and capitals, verb tenses, and other relationships - without explicit supervision!

This wasn't just a party trick; it demonstrated that these models captured meaningful semantic and syntactic relationships. Researchers were astonished that such rich structure emerged from simple prediction tasks.

This discovery fundamentally changed NLP's approach to meaning representation and sparked intense interest in what other linguistic knowledge might be implicitly encoded in neural representations.

### 6.3 The Distributional Hypothesis

The theoretical foundation for word embeddings comes from linguistics, specifically the **distributional hypothesis**, which states:

> "You shall know a word by the company it keeps." - J.R. Firth (1957)

This elegant idea suggests that words appearing in similar contexts tend to have similar meanings. For example, "dog" and "cat" often appear in similar contexts: both can be pets, both can be fed, walked, etc.

#### From Hypothesis to Algorithm

Word embedding algorithms operationalize this hypothesis by:

1. Collecting statistics about word co-occurrences in large text corpora
2. Constructing word vectors such that words occurring in similar contexts have similar vectors
3. Optimizing these vectors to predict a word's context or vice versa

#### Context Defines Meaning

Consider these sentences:
- "The ____ is barking loudly."
- "The ____ is purring on the sofa."

Without knowing the missing words, we can guess that the first blank likely refers to a dog or similar animal, while the second probably refers to a cat. This intuition—that context helps determine meaning—is precisely what distributed word representations capture mathematically.

### 6.4 Properties of Word Embeddings

Word embeddings have several remarkable properties that make them powerful tools for NLP tasks:

#### 1. Semantic Similarity

Words with similar meanings have vectors that are close in the vector space. This allows us to measure semantic similarity using vector distance metrics like cosine similarity:

$\cos(\theta) = \frac{\vec{a} \cdot \vec{b}}{||\vec{a}|| \cdot ||\vec{b}||}$

The closer this value is to 1, the more similar the words.

#### 2. Linear Substructures

Perhaps the most fascinating property is that linguistic relationships are captured as vector offsets. For example:

$\vec{king} - \vec{man} + \vec{woman} \approx \vec{queen}$

Other examples include:
- $\vec{Paris} - \vec{France} + \vec{Italy} \approx \vec{Rome}$
- $\vec{walked} - \vec{walking} + \vec{running} \approx \vec{ran}$

These linear relationships emerge organically from the training process!

#### 3. Dimensionality Reduction

Word embeddings are far more compact than one-hot encodings, typically using 50-300 dimensions regardless of vocabulary size. This compression retains semantic information while making computations much more efficient.

#### 4. Transferability

Embeddings trained on one large corpus can be used for multiple downstream tasks. This is an early example of transfer learning in NLP, where knowledge gained from one task (predicting context) transfers to other tasks.

### Aside: The Dimensionality Question

How many dimensions should word embeddings have? This seemingly technical question has profound implications.

Early models used 50-300 dimensions, balancing expressivity against computational constraints. But what does each dimension represent? Unlike image features, embedding dimensions lack clear interpretability—they're emergent properties of the training process.

Some researchers have attempted to interpret them post-hoc, finding dimensions that correlate with concepts like gender, plurality, or sentiment, but most remain opaque. Techniques like Principal Component Analysis can sometimes reveal that certain directions in the embedding space correspond to interpretable concepts.

The number of dimensions represents a hypothesis about language's inherent complexity—too few dimensions can't capture nuanced relationships, while too many lead to overfitting and wasted computation. It's like deciding how many colors you need to paint a realistic picture: 16 colors might be too few to capture subtle shades, while 16 million might be unnecessarily precise.

Interestingly, modern contextualized embeddings use much higher dimensions (768-4096), suggesting our earlier models may have been underpowered. This dimensional expansion reflects our growing understanding of the complexity of language and its representation.

### 6.5 Evaluation Methods for Word Embeddings

How do we know if word embeddings are "good"? This question has led to various evaluation methods that test different aspects of embedding quality.

#### Intrinsic Evaluation

Intrinsic methods directly test properties of the embeddings themselves:

1. **Word Similarity Tasks**:
   - Datasets: WordSim-353, SimLex-999, MEN
   - Method: Compare human similarity judgments with cosine similarities
   - Metrics: Spearman or Pearson correlation between human and embedding rankings

2. **Word Analogy Tasks**:
   - Datasets: Google Analogy, SemEval
   - Method: Test vector arithmetic like "king - man + woman ≈ queen"
   - Metrics: Accuracy in predicting the correct word

3. **Concept Categorization**:
   - Datasets: BLESS, ESSLLI
   - Method: Cluster word vectors and compare with human categorizations
   - Metrics: Purity, normalized mutual information

#### Extrinsic Evaluation

Extrinsic methods evaluate embeddings by their performance on downstream NLP tasks:

1. **Text Classification**
2. **Named Entity Recognition**
3. **Sentiment Analysis**
4. **Part-of-Speech Tagging**

These evaluations offer a more practical measure of embedding quality for specific applications.

## Section 7: Word2Vec: Skip-gram Model

In our journey through word embedding techniques, we've now arrived at one of the most influential algorithms in NLP: Word2Vec. This section focuses on the Skip-gram architecture, which revolutionized how we represent words as vectors.

Recall that in the previous section, we discussed the concept of distributed word representations and why they're superior to one-hot encodings. Word2Vec provides an elegant method to learn these distributed representations from text data, capturing semantic and syntactic relationships between words in a remarkably efficient manner.

The Skip-gram model approaches the embedding problem with a simple but powerful idea: predict the context words that appear around a given target word. By learning to make these predictions, the model develops vector representations that encode meaningful relationships between words.

### 7.1 Skip-gram Architecture and Training Objective

The Skip-gram model has a deceptively simple architecture with a profound learning objective. Unlike traditional language models that predict the next word given previous words, Skip-gram does something quite different.

#### The Basic Idea

Given a target word, Skip-gram attempts to predict the surrounding context words within a window. For example, if our sentence is "The quick brown fox jumps" and our target word is "brown" with a window size of 2, the model would try to predict "The", "quick", "fox", and "jumps" from "brown".

#### Mathematical Formulation

Formally, Skip-gram maximizes the average log probability of context words given a target word:

$$J(\theta) = \frac{1}{T}\sum_{t=1}^{T}\sum_{-c \leq j \leq c, j \neq 0} \log p(w_{t+j}|w_t)$$

where:
- $T$ is the number of words in the corpus
- $c$ is the context window size
- $w_t$ is the target word
- $w_{t+j}$ is a context word
- $p(w_{t+j}|w_t)$ is modeled using the softmax function:

$$p(w_O|w_I) = \frac{\exp(\mathbf{v'}_{w_O}^\top \mathbf{v}_{w_I})}{\sum_{w=1}^{V} \exp(\mathbf{v'}_w^\top \mathbf{v}_{w_I})}$$

where:
- $\mathbf{v}_{w_I}$ is the input vector (embedding) of the target word
- $\mathbf{v'}_{w_O}$ is the output vector of the context word
- $V$ is the vocabulary size

This formulation has a computational challenge: the denominator requires summing over the entire vocabulary (often 100,000+ words), making it prohibitively expensive to compute for each training example.

#### Architectural Details

The Skip-gram model consists of three main components:

1. **Input Layer**: A one-hot encoded vector representing the target word
2. **Hidden Layer**: A linear projection layer that maps the one-hot vector to a dense embedding space (no activation function)
3. **Output Layer**: A softmax layer that outputs probabilities for each word in the vocabulary being a context word

The weights between the input and hidden layer form the word embedding matrix we're interested in. Each row corresponds to a word's embedding vector.

Interestingly, there are actually two sets of embeddings being learned:
- Input embeddings for target words
- Output embeddings for context words

After training, typically only the input embeddings are used for downstream tasks, though some approaches average the two.

### Aside: The One-Sentence Insight

The brilliance of Skip-gram can be distilled to one key insight: "A good word embedding should be good at predicting its neighbors." This simple objective produces vectors that capture deep semantic relationships. Think about it - words that appear in similar contexts should have similar meanings. By forcing the model to predict context from words, we're essentially encoding the distributional hypothesis ("you shall know a word by the company it keeps") directly into our optimization objective. This also explains why Skip-gram works well with relatively small datasets - the training objective creates multiple training examples from each word occurrence, as each word must predict several context words.

### 7.2 Negative Sampling

As mentioned earlier, the full softmax computation for the Skip-gram objective is prohibitively expensive due to the normalization term that sums over the entire vocabulary. Negative Sampling is an efficient approximation that transforms the problem from a V-way classification (where V is vocabulary size) into multiple binary classifications.

#### The Negative Sampling Objective

Instead of predicting the probability distribution over all words in the vocabulary, Negative Sampling focuses on distinguishing the actual context words from randomly sampled "negative" words. The new objective function becomes:

$$\log \sigma(\mathbf{v'}_{w_O}^\top \mathbf{v}_{w_I}) + \sum_{i=1}^{k} \mathbb{E}_{w_i \sim P_n(w)} [\log \sigma(-\mathbf{v'}_{w_i}^\top \mathbf{v}_{w_I})]$$

where:
- $\sigma$ is the sigmoid function $\sigma(x) = \frac{1}{1+e^{-x}}$
- $k$ is the number of negative samples (typically 5-20)
- $P_n(w)$ is the noise distribution for sampling negative examples
- $w_O$ is the actual context word (positive example)
- $w_i$ are randomly sampled words (negative examples)

#### Selecting Negative Samples

The choice of noise distribution $P_n(w)$ is important for effective training. The authors found that using the unigram distribution raised to the 3/4 power works better than a uniform or unmodified unigram distribution:

$$P_n(w) = \frac{f(w)^{3/4}}{\sum_{i=1}^{V}f(i)^{3/4}}$$

where $f(w)$ is the frequency of word $w$ in the corpus.

This modification gives less weight to very frequent words and more weight to rare words compared to the raw frequency, creating a better balance for negative samples.

#### Why Negative Sampling Works

Negative sampling approximates the full softmax by only updating a small subset of weights in each training step. For each positive context word, we update:
- The vector for the target word
- The vector for the actual context word (pushing these closer together)
- The vectors for k negative samples (pushing these further from the target word)

This is much more efficient than updating all V output vectors in the standard softmax approach.

````python
### Code Example: Skip-gram with Negative Sampling

import torch
import torch.nn as nn
import torch.optim as optim
import numpy as np
from collections import Counter
import random

class SkipGramNegativeSampling(nn.Module):
    def __init__(self, vocab_size, embedding_dim):
        super(SkipGramNegativeSampling, self).__init__()
        
        # Embedding layers for input and output words
        # We learn two embeddings: one for target words and one for context words
        self.target_embeddings = nn.Embedding(vocab_size, embedding_dim)
        self.context_embeddings = nn.Embedding(vocab_size, embedding_dim)
        
        # Initialize embeddings with small random values
        self.target_embeddings.weight.data.uniform_(-0.1, 0.1)
        self.context_embeddings.weight.data.uniform_(-0.1, 0.1)
        
    def forward(self, target_word, context_word, negative_samples):
        # Get embeddings for target and positive context words
        target_emb = self.target_embeddings(target_word)  # [batch_size, embed_dim]
        context_emb = self.context_embeddings(context_word)  # [batch_size, embed_dim]
        neg_emb = self.context_embeddings(negative_samples)  # [batch_size, num_neg_samples, embed_dim]
        
        # Positive pair score
        positive_score = torch.sum(target_emb * context_emb, dim=1)  # [batch_size]
        positive_score = torch.clamp(torch.sigmoid(positive_score), min=1e-8, max=1-1e-8)  # Numerical stability
        positive_loss = -torch.log(positive_score)  # [batch_size]
        
        # Negative pairs score
        # Broadcasting: target_emb [batch_size, 1, embed_dim] x neg_emb [batch_size, num_neg, embed_dim]
        target_emb_expanded = target_emb.unsqueeze(1)
        neg_score = torch.bmm(neg_emb, target_emb_expanded.transpose(1, 2)).squeeze(2)  # [batch_size, num_neg]
        neg_score = torch.clamp(torch.sigmoid(-neg_score), min=1e-8, max=1-1e-8)  # Numerical stability
        negative_loss = -torch.sum(torch.log(neg_score), dim=1)  # [batch_size]
        
        loss = positive_loss + negative_loss
        return loss.mean()

# Example usage
vocab_size = 10000
embedding_dim = 300
model = SkipGramNegativeSampling(vocab_size, embedding_dim)

# Sample data (normally you would have a proper data loader)
target_word = torch.LongTensor([1, 2, 3])  # Batch of 3 target word indices
context_word = torch.LongTensor([4, 5, 6])  # Their respective context word indices
negative_samples = torch.LongTensor([  # 5 negative samples for each target word
    [10, 20, 30, 40, 50],
    [11, 21, 31, 41, 51],
    [12, 22, 32, 42, 52]
])

# Calculate loss
loss = model(target_word, context_word, negative_samples)
print(f"Loss: {loss.item()}")

# In practice, you would then call loss.backward() and update with an optimizer
````

The negative sampling technique revolutionized word embedding training by solving a computational bottleneck. The original softmax objective required summing over the entire vocabulary (often 100,000+ words) for each prediction - computationally prohibitive. Instead, negative sampling turns the problem into binary classification: distinguish true context pairs from randomly sampled 'negative' pairs. With just 5-20 negative samples per positive example, training becomes feasible on ordinary hardware. The sampling distribution matters too - Word2Vec uses the unigram distribution raised to the 3/4 power, empirically found to work better than uniform or raw frequency sampling. This seemingly minor optimization detail enabled the word embedding revolution by making training practical at scale.

### 7.3 Hierarchical Softmax

Hierarchical Softmax is an alternative to Negative Sampling for efficiently training Skip-gram models. While Negative Sampling approximates the softmax by sampling a small subset of negative examples, Hierarchical Softmax represents the vocabulary as a binary tree, reducing the computational complexity from O(V) to O(log V).

#### Binary Tree Representation

In Hierarchical Softmax, words are represented as leaf nodes in a binary tree. The probability of a context word given a target word is defined as the probability of traversing the path from the root of the tree to the leaf node corresponding to the context word.

````python
### Hierarchical Softmax Visualization

from IPython.display import Image
import requests
import io

try:
    # Try to load the image from URL
    url = "https://miro.medium.com/v2/resize:fit:1400/1*YEJhKxRHr6OAJJn-5W9d6Q.png"
    response = requests.get(url)
    img_data = io.BytesIO(response.content)
    display(Image(data=img_data.getvalue(), width=600))
    print("Hierarchical softmax tree representation: Words are leaf nodes and the probability")
    print("of a context word is calculated by traversing the path from root to leaf")
except:
    print("Failed to load image. Here's what it would show:")
    print("A binary tree structure with words as leaf nodes, and internal nodes as binary classifiers.")
    print("The path from root to a leaf represents a series of binary decisions,")
    print("effectively decomposing the large softmax computation into log(V) binary decisions.")
````

#### Mathematical Formulation

For a given target word $w_I$ and context word $w_O$, the probability is calculated as:

$$p(w_O|w_I) = \prod_{j=1}^{L(w_O)-1} \sigma([n(w_O, j+1) = \text{left child of } n(w_O, j)] \cdot \mathbf{v}_{n(w_O, j)}^\top\mathbf{v}_{w_I})$$

where:
- $L(w_O)$ is the length of the path from the root to the leaf node $w_O$
- $\mathbf{n}(w_O, j)$ is the j-th node on the path
- $\sigma$ is the sigmoid function
- $[x]$ is 1 if x is true and -1 otherwise

Each internal node effectively acts as a binary classifier deciding which sub-tree to navigate to.

#### Building an Efficient Tree

The structure of the tree significantly impacts efficiency. Two common approaches are:

1. **Balanced Binary Tree**: Ensures log(V) depth for all words, but doesn't leverage word frequency information

2. **Huffman Tree**: Creates a binary tree based on word frequencies, assigning shorter paths to more frequent words, which further improves efficiency

#### Hierarchical Softmax vs. Negative Sampling

Both techniques aim to make Skip-gram training efficient, but they have different characteristics:

- **Hierarchical Softmax**:
  - Better for rare words (all words are updated in each step)
  - More consistent probability estimates
  - Computationally predictable (always log(V) operations)

- **Negative Sampling**:
  - Easier to implement
  - Often faster in practice with very large vocabularies
  - Better for frequent words and smaller embedding dimensions
  - Focuses computational resources on more relevant examples

In practice, Negative Sampling has become more widely used due to its simplicity and empirical performance, but Hierarchical Softmax remains an important technique, especially when accurate probability estimates are needed.

### 7.4 Hyperparameter Considerations

The performance of Skip-gram models depends significantly on several hyperparameters. Understanding these can help you train more effective word embeddings for your specific tasks.

#### Key Hyperparameters

1. **Embedding Dimension**
   - Typically ranges from 50 to 300
   - Larger dimensions can capture more information but require more data and computation
   - Smaller dimensions (50-100) work well for syntactic tasks
   - Larger dimensions (200-300) better for semantic tasks

2. **Context Window Size**
   - Defines how many words before and after the target word to consider as context
   - Smaller windows (2-5): Better for capturing syntactic relationships and functional similarity
   - Larger windows (5-10+): Better for semantic relationships and topical similarity
   - Example: "The quick brown fox jumps over the lazy dog"
     - With window size 2 for target "brown": context is ["The", "quick", "fox", "jumps"]

3. **Negative Samples**
   - Number of negative examples per positive context word
   - Typical range: 5-20 negatives per positive context
   - Fewer samples (2-5) for larger datasets
   - More samples (10-15) for smaller datasets

4. **Subsampling Threshold**
   - Used to downsample very frequent words (like "the", "and")
   - Words are randomly discarded with probability:
   $$P(w_i) = 1 - \sqrt{\frac{t}{f(w_i)}}$$
   - where $t$ is the threshold (typically 10^-5) and $f(w_i)$ is the frequency
   - Improves both training speed and quality of representations for less frequent words

````python
### Exercise: Hyperparameter Experimentation

# Let's create a function to visualize how different window sizes affect
# the training examples generated from a sentence

def generate_skip_gram_examples(sentence, window_sizes=[1, 2, 3]):
    """Generate Skip-gram training pairs for different window sizes"""
    words = sentence.lower().split()
    examples = {}
    
    for window_size in window_sizes:
        examples[window_size] = []
        
        for i, target in enumerate(words):
            # For each position, get context words within the window
            left_context = words[max(0, i - window_size):i]
            right_context = words[i+1:min(i + window_size + 1, len(words))]
            context = left_context + right_context
            
            # Add (target, context) pairs
            for ctx_word in context:
                examples[window_size].append((target, ctx_word))
    
    return examples

# Test with a sample sentence
sample_sentence = "The quick brown fox jumps over the lazy dog"
window_examples = generate_skip_gram_examples(sample_sentence, [1, 2, 3])

# Display the examples
for window, pairs in window_examples.items():
    print(f"Window size {window} generates {len(pairs)} training examples:")
    for target, context in pairs:
        print(f"  Target: '{target}', Context: '{context}'")
    print()

# Challenge: Modify the function to implement subsampling of frequent words
# Hint: You'll need a corpus to calculate word frequencies first
````

### Aside: The Window Size Impact

The context window size in Skip-gram is a crucial hyperparameter that dramatically affects what the embeddings learn. Small windows (2-5 words) tend to capture syntactic and functional similarities - words with similar roles in sentences group together regardless of topic. Larger windows (5-10+ words) capture more topical and semantic similarities - words related to similar concepts cluster together. This isn't just theoretical; it affects downstream applications. ### 7.5 Skip-gram Implementation and Training

#### Data Preparation

Before training, we need to prepare our data by:
1. Tokenizing the text
2. Building a vocabulary (with frequency counts)
3. Generating training pairs (target word, context word)
4. Implementing negative sampling

#### Training Process Overview

The training process involves:
1. Batching training pairs
2. Forward pass to compute loss with negative sampling
3. Backward pass to update embeddings
4. Periodic evaluation on similarity or analogy tasks

Let's implement the core components of this pipeline:

````python
### Code: Skip-gram Training Pipeline

import numpy as np
import torch
import torch.nn as nn
import torch.optim as optim
from collections import Counter
import random

# 1. Data preparation functions
def preprocess_text(text, min_count=5):
    """Tokenize text and build vocabulary"""
    # Simple tokenization by splitting on whitespace
    words = text.lower().split()
    
    # Count word frequencies
    word_counts = Counter(words)
    
    # Filter words below minimum count
    word_counts = {w: c for w, c in word_counts.items() if c >= min_count}
    
    # Create word-to-index mappings
    word_to_idx = {w: i for i, w in enumerate(word_counts.keys())}
    idx_to_word = {i: w for w, i in word_to_idx.items()}
    
    # Convert text to indices, skipping rare words
    word_indices = [word_to_idx[w] for w in words if w in word_to_idx]
    
    return word_indices, word_to_idx, idx_to_word, word_counts

def generate_training_data(word_indices, word_counts, window_size=2):
    """Generate training pairs of (target, context) words"""
    training_pairs = []
    
    # Calculate total frequency for negative sampling
    total_count = sum([count**0.75 for count in word_counts.values()])
    word_freqs = {word: (count**0.75)/total_count for word, count in word_counts.items()}
    
    # Create sampling table for negative sampling
    sampling_table = []
    for word, freq in word_freqs.items():
        sampling_table.extend([word] * int(freq * 1e8))  # Scale up to avoid floating point issues
    
    # Generate training pairs
    for i, target_idx in enumerate(word_indices):
        # Define context window
        window_start = max(0, i - window_size)
        window_end = min(len(word_indices), i + window_size + 1)
        
        # For each context position
        for j in range(window_start, window_end):
            if j != i:  # Skip the target word itself
                context_idx = word_indices[j]
                training_pairs.append((target_idx, context_idx))
    
    return training_pairs, sampling_table

# 2. Skip-gram model with negative sampling
class SkipGramModel(nn.Module):
    def __init__(self, vocab_size, embedding_dim):
        super(SkipGramModel, self).__init__()
        self.target_embeddings = nn.Embedding(vocab_size, embedding_dim)
        self.context_embeddings = nn.Embedding(vocab_size, embedding_dim)
        
        # Initialize embeddings
        self.target_embeddings.weight.data.uniform_(-0.1, 0.1)
        self.context_embeddings.weight.data.uniform_(-0.1, 0.1)
    
    def forward(self, target_idx, context_idx, neg_context_idx):
        target_emb = self.target_embeddings(target_idx)  # [batch_size, embed_dim]
        context_emb = self.context_embeddings(context_idx)  # [batch_size, embed_dim]
        neg_emb = self.context_embeddings(neg_context_idx)  # [batch_size, num_neg, embed_dim]
        
        # Positive score
        pos_score = torch.sum(target_emb * context_emb, dim=1)  # [batch_size]
        pos_score = torch.sigmoid(pos_score)  # [batch_size]
        
        # Negative score
        target_emb = target_emb.unsqueeze(1)  # [batch_size, 1, embed_dim]
        neg_score = torch.bmm(neg_emb, target_emb.transpose(1, 2)).squeeze()  # [batch_size, num_neg]
        neg_score = torch.sigmoid(-neg_score)  # [batch_size, num_neg]
        
        # Loss calculation
        pos_loss = -torch.mean(torch.log(pos_score + 1e-10))
        neg_loss = -torch.mean(torch.log(neg_score + 1e-10))
        
        return pos_loss + neg_loss
    
    def get_target_embeddings(self):
        return self.target_embeddings.weight.data.cpu().numpy()

# 3. Training function
def train_skip_gram(model, training_pairs, sampling_table, batch_size=128, n_negatives=5, epochs=5, lr=0.01):
    optimizer = optim.Adam(model.parameters(), lr=lr)
    scheduler = optim.lr_scheduler.StepLR(optimizer, step_size=1, gamma=0.9)
    
    # Convert to numpy array for faster indexing
    sampling_table = np.array(sampling_table)
    
    for epoch in range(epochs):
        # Shuffle training pairs
        random.shuffle(training_pairs)
        
        total_loss = 0
        batch_count = 0
        
        for i in range(0, len(training_pairs), batch_size):
            # Get batch of training pairs
            batch_pairs = training_pairs[i:i+batch_size]
            if not batch_pairs:
                continue
                
            targets, contexts = zip(*batch_pairs)
            targets = torch.LongTensor(targets)
            contexts = torch.LongTensor(contexts)
            
            # Sample negative contexts
            neg_contexts = []
            for target in targets.numpy():  # Get target words for the batch
                negatives = []
                while len(negatives) < n_negatives:
                    neg = random.choice(sampling_table)
                    if neg != target:  # Don't use the target word as a negative
                        negatives.append(neg)
                neg_contexts.append(negatives)
            neg_contexts = torch.LongTensor(neg_contexts)
            
            # Forward and backward passes
            optimizer.zero_grad()
            loss = model(targets, contexts, neg_contexts)
            loss.backward()
            optimizer.step()
            
            total_loss += loss.item()
            batch_count += 1
            
            # Print progress
            if batch_count % 1000 == 0:
                print(f"Epoch {epoch+1}, Batch {batch_count}, Loss: {total_loss/batch_count:.4f}")
        
        # Decay learning rate
        scheduler.step()
        print(f"Epoch {epoch+1} completed. Average loss: {total_loss/batch_count:.4f}")
    
    return model

# Example usage with a small sample text
sample_text = """Natural language processing has evolved significantly with the advent of word embeddings. 
               Word2Vec models like Skip-gram learn distributed representations of words based on their context. 
               These embeddings capture semantic and syntactic relationships between words.
               Skip-gram works by predicting context words given a target word."""

print("This is a demo with a tiny corpus. In practice, you would need millions of words for good embeddings.\n")

# Preprocess text
word_indices, word_to_idx, idx_to_word, word_counts = preprocess_text(sample_text, min_count=1)
print(f"Vocabulary size: {len(word_to_idx)}")

# Generate training pairs
training_pairs, sampling_table = generate_training_data(word_indices, word_counts, window_size=2)
print(f"Generated {len(training_pairs)} training pairs with window size 2")

# Create and train model (just 2 epochs for demonstration)
vocab_size = len(word_to_idx)
embedding_dim = 10  # Small dimension for demo
model = SkipGramModel(vocab_size, embedding_dim)
trained_model = train_skip_gram(model, training_pairs, sampling_table, 
                               batch_size=4, n_negatives=2, epochs=2, lr=0.01)

# Get embeddings for a few example words
embeddings = trained_model.get_target_embeddings()
for word in ['word', 'embeddings', 'skip', 'gram']:
    if word in word_to_idx:
        idx = word_to_idx[word]
        print(f"\nEmbedding for '{word}': {embeddings[idx][:3]}... (showing first 3 dimensions)")
````

#### Evaluating the Embeddings

After training, we should evaluate our embeddings to ensure they've captured meaningful relationships. Common evaluation methods include:

1. **Word Similarity Tasks**
   - Compare cosine similarity of word pairs with human judgments
   - Datasets: WordSim353, SimLex-999, MEN

2. **Word Analogy Tasks**
   - Test if embeddings can solve analogies like "man is to woman as king is to queen"
   - Computed as argmax_w(cos(w, king - man + woman))
   - Google analogy dataset contains semantic and syntactic analogies

3. **Visualization**
   - Use dimensionality reduction (t-SNE, UMAP) to visualize embeddings in 2D
   - Check if semantically related words cluster together

````python
### Code: Evaluating Word Embeddings

import numpy as np
import matplotlib.pyplot as plt

def cosine_similarity(v1, v2):
    """Compute cosine similarity between two vectors"""
    return np.dot(v1, v2) / (np.linalg.norm(v1) * np.linalg.norm(v2))

def find_similar_words(word, word_to_idx, idx_to_word, embeddings, top_k=5):
    """Find most similar words to a query word"""
    if word not in word_to_idx:
        return []
    
    word_idx = word_to_idx[word]
    word_vec = embeddings[word_idx]
    
    # Compute similarities to all other words
    similarities = []
    for i, embedding in enumerate(embeddings):
        if i != word_idx:
            similarity = cosine_similarity(word_vec, embedding)
            similarities.append((idx_to_word[i], similarity))
    
    # Sort by similarity and take top k
    return sorted(similarities, key=lambda x: x[1], reverse=True)[:top_k]

def solve_analogy(word1, word2, word3, word_to_idx, idx_to_word, embeddings, top_k=5):
    """Solve word analogy: word1 is to word2 as word3 is to ?"""
    if not all(w in word_to_idx for w in [word1, word2, word3]):
        return []
    
    # Get word vectors
    vec1 = embeddings[word_to_idx[word1]]
    vec2 = embeddings[word_to_idx[word2]]
    vec3 = embeddings[word_to_idx[word3]]
    
    # Calculate target vector: vec2 - vec1 + vec3
    target_vec = vec2 - vec1 + vec3
    
    # Find closest words to target vector
    similarities = []
    for i, embedding in enumerate(embeddings):
        # Skip the query words
        if idx_to_word[i] not in [word1, word2, word3]:
            similarity = cosine_similarity(target_vec, embedding)
            similarities.append((idx_to_word[i], similarity))
    
    return sorted(similarities, key=lambda x: x[1], reverse=True)[:top_k]

# Example usage (with our tiny model)
print("Finding similar words (note: our toy model is too small for meaningful results)")
try:
    similar_to_word = find_similar_words('word', word_to_idx, idx_to_word, embeddings)
    print("\nWords similar to 'word':")
    for word, similarity in similar_to_word:
        print(f"  {word}: {similarity:.4f}")
    
    print("\nWith a properly trained model on a large corpus, we could perform analogy tasks like:")
    print("  man is to woman as king is to queen")
    print("  paris is to france as berlin is to germany")
    print("  walking is to walked as swimming is to swam")
except:
    print("Need more data for meaningful evaluation!")

print("\nTo visualize embeddings with t-SNE or UMAP, we would need:")
print("1. Larger embedding dimensions (typically 100-300)")
print("2. More training data (millions of words)")
print("3. More training epochs")
````

### 8.1 CBOW Architecture and Objective

The Continuous Bag of Words (CBOW) model predicts a target word based on its context - essentially asking "what word is likely to appear in this context?"

#### Architecture

CBOW takes multiple context words as input and predicts a single target word:

1. **Input**: Context words surrounding the target (e.g., 2 words before and 2 words after)
2. **Projection**: Average the context word vectors into a single representation
3. **Output**: Predict the probability distribution over the vocabulary for the target word

![CBOW Architecture](https://i.imgur.com/qspE9Ar.png)

#### Training Objective

For a specific target word $w_t$ and context words $c_{t-m},...,c_{t-1},c_{t+1},...,c_{t+m}$, CBOW maximizes the probability of the target given its context:

$$P(w_t | c_{t-m},...,c_{t-1},c_{t+1},...,c_{t+m})$$

This is typically formulated as a softmax over the vocabulary:

$$P(w_t|context) = \frac{\exp(v_{w_t}^{\top} \cdot \hat{v}_{context})}{\sum_{w \in V} \exp(v_w^{\top} \cdot \hat{v}_{context})}$$

Where $\hat{v}_{context}$ is the average of the context word vectors:

$$\hat{v}_{context} = \frac{1}{2m} \sum_{-m \leq j \leq m, j \neq 0} v_{c_{t+j}}$$

### Aside: The Etymology of "Continuous Bag of Words"

The name "Continuous Bag of Words" might seem peculiar, especially the "continuous" part. This name originated to distinguish it from traditional bag-of-words models in NLP:

- **Traditional Bag of Words**: Represents text as sparse count vectors showing which words appear, ignoring order
- **Continuous Bag of Words**: Uses "continuous" vector space representations instead of discrete counts

The "bag" refers to the fact that context words are averaged, losing their order information, similar to how a traditional bag-of-words model discards word order.

Mikolov et al. chose this name to emphasize that while the model discards word order within the context window (like a bag), it uses dense continuous vectors unlike traditional sparse representations. The name stuck, though some researchers joke it should have been called "context-to-word prediction model" for clarity!

### 8.2 Comparison with Skip-gram

CBOW and Skip-gram are mirror images of each other, but this inversion leads to significant differences in what they learn and how they perform.

#### Architectural Comparison

| Feature | CBOW | Skip-gram |
| --- | --- | --- |
| **Prediction task** | Predict target from context | Predict context from target |
| **Direction** | Many-to-one | One-to-many |
| **Input** | Average of context vectors | Target word vector |
| **Training examples** | One per target word | Multiple (one per context position) |

![CBOW vs Skip-gram](https://i.imgur.com/ADC1szU.png)

#### Performance Comparison

| Aspect | CBOW | Skip-gram |
| --- | --- | --- |
| **Training speed** | Faster (2-3x) | Slower |
| **Frequent words** | Performs better | Less emphasis |
| **Rare words** | Less effective | Performs better |
| **Embedding smoothness** | Smoother (due to averaging) | More specific |
| **Syntactic vs. semantic** | Better at syntax | Better at semantics |

The fundamental trade-off: CBOW's vector averaging smooths out noise, making it better for frequent words and syntactic relationships, while Skip-gram's multiple training examples per target word help it better capture rare words and semantic relationships.

### 8.3 Implementation Details

Let's implement the CBOW model from scratch to understand its inner workings. The key steps are:

1. Creating training pairs (context -> target)
2. Building the model architecture
3. Implementing negative sampling for efficient training

#### Training Data Preparation

Unlike Skip-gram, which creates multiple context pairs per target word, CBOW creates a single training example for each target word by bundling its surrounding context words.

For example, with the sentence "The quick brown fox jumps over the lazy dog" and a window size of 2, we would generate these training pairs:

- Context: ["the", "brown", "jumps"] → Target: "fox"
- Context: ["quick", "fox", "over"] → Target: "jumps"

The input is the average of the context word vectors, and the output is the target word.

````python
### CBOW Data Preparation

import numpy as np
import torch

def create_cbow_dataset(text, window_size=2, min_count=1):
    """Creates CBOW training pairs from text corpus."""
    # Tokenize the text
    words = text.lower().split()
    
    # Create vocabulary
    word_counts = {}
    for word in words:
        word_counts[word] = word_counts.get(word, 0) + 1
    
    # Filter vocabulary by frequency
    vocab = {word: idx for idx, (word, count) in enumerate(
        filter(lambda x: x[1] >= min_count, word_counts.items()))
    }
    
    # Create inverted vocabulary mapping
    idx_to_word = {idx: word for word, idx in vocab.items()}
    
    # Create context-target pairs
    context_pairs = []
    # Convert words to indices, ignoring OOV words
    word_indices = [vocab[word] for word in words if word in vocab]
    
    for i in range(len(word_indices)):
        # Get the target word
        target = word_indices[i]
        
        # Define the window boundaries
        start = max(0, i - window_size)
        end = min(len(word_indices), i + window_size + 1)
        
        # Create context by excluding the target
        context = [word_indices[j] for j in range(start, end) if j != i]
        
        # Add the context-target pair
        if len(context) > 0:  # Ensure we have at least one context word
            context_pairs.append((context, target))
    
    return context_pairs, vocab, idx_to_word

# Example usage
text = """The quick brown fox jumps over the lazy dog. 
        The dog wakes up and follows the fox into the forest."""

pairs, vocab, idx_to_word = create_cbow_dataset(text, window_size=2)

# Display first few training pairs
for i, (context, target) in enumerate(pairs[:5]):
    context_words = [idx_to_word[idx] for idx in context]
    print(f"Pair {i+1}: Context {context_words} → Target '{idx_to_word[target]}'")
````

#### CBOW Model Architecture

Now let's implement the CBOW model architecture in PyTorch. The key components are:

1. Input layer for embedding context words
2. Projection layer that averages context embeddings
3. Output layer for predicting the target word

Like Skip-gram, we'll use negative sampling for efficient training.

````python
### CBOW Model Implementation with PyTorch

import torch
import torch.nn as nn
import torch.optim as optim
import numpy as np

class CBOW(nn.Module):
    def __init__(self, vocab_size, embedding_dim):
        super(CBOW, self).__init__()
        
        # Embedding layers for input and output
        # These are the same as in Skip-gram, but used differently
        self.embeddings = nn.Embedding(vocab_size, embedding_dim)
        self.output_embeddings = nn.Embedding(vocab_size, embedding_dim)
        
        # Initialize weights
        self.init_weights()
        
    def init_weights(self):
        # Initialize embeddings from a uniform distribution
        nn.init.uniform_(self.embeddings.weight, -0.1, 0.1)
        nn.init.uniform_(self.output_embeddings.weight, -0.1, 0.1)
    
    def forward(self, contexts, targets, negative_samples):
        # contexts shape: [batch_size, variable_context_size]
        # targets shape: [batch_size]
        # negative_samples shape: [batch_size, num_negative]
        
        # Get embeddings for all context words
        context_embeds = self.embeddings(contexts)  # [batch_size, context_size, embed_dim]
        
        # Average the context embeddings
        # Need to handle variable context sizes, so we compute mean along dim=1
        context_mean = torch.mean(context_embeds, dim=1)  # [batch_size, embed_dim]
        
        # Get target and negative sample embeddings
        target_embeds = self.output_embeddings(targets)  # [batch_size, embed_dim]
        neg_embeds = self.output_embeddings(negative_samples)  # [batch_size, num_neg, embed_dim]
        
        # Compute positive score (dot product of context mean and target)
        pos_score = torch.sum(context_mean * target_embeds, dim=1)  # [batch_size]
        pos_score = torch.clamp(torch.sigmoid(pos_score), min=1e-8, max=1-1e-8)
        
        # Compute negative scores
        # First reshape context_mean to [batch_size, 1, embed_dim] for broadcasting
        neg_score = torch.bmm(neg_embeds, context_mean.unsqueeze(2)).squeeze(2)  # [batch_size, num_neg]
        neg_score = torch.clamp(torch.sigmoid(-neg_score), min=1e-8, max=1-1e-8)
        
        # Compute loss
        # Negative log likelihood for positive and negative samples
        pos_loss = -torch.mean(torch.log(pos_score))
        neg_loss = -torch.mean(torch.sum(torch.log(neg_score), dim=1))
        
        return pos_loss + neg_loss
    
    def get_word_embeddings(self):
        # For most applications, we use the input embeddings as our word vectors
        return self.embeddings.weight.detach().numpy()

# Example of how to prepare variable-length context tensors
def collate_cbow_batch(batch, vocab_size, num_negatives=5):
    """Collate function for CBOW that handles variable context sizes"""
    # Unzip the batch
    contexts, targets = zip(*batch)
    
    # Create padded context tensor and mask
    max_len = max(len(context) for context in contexts)
    padded_contexts = []
    
    for context in contexts:
        # Handle variable-length contexts by padding
        padded = context + [0] * (max_len - len(context))
        padded_contexts.append(padded)
    
    # Convert to tensors
    contexts_tensor = torch.LongTensor(padded_contexts)
    targets_tensor = torch.LongTensor(targets)
    
    # Generate negative samples
    batch_size = len(targets)
    # In practice, you'd want to sample based on word frequency
    neg_samples = torch.randint(1, vocab_size, (batch_size, num_negatives))
    
    return contexts_tensor, targets_tensor, neg_samples

# Demonstration of model usage (not actual training)
vocab_size = len(vocab)
embedding_dim = 50
model = CBOW(vocab_size, embedding_dim)
print(f"Created CBOW model with vocabulary size {vocab_size} and embedding dimension {embedding_dim}")
````

### 8.4 Training and Optimization

Training a CBOW model involves several important considerations that influence performance and efficiency:

#### Training Process

1. **Data Preparation**: Create context-target pairs from the corpus
2. **Batching**: Group similar-length contexts for efficient processing
3. **Negative Sampling**: Sample negative examples for each positive pair
4. **Optimization**: Update embeddings using gradients from the loss function

#### Optimization Strategies

- **Learning Rate Schedule**: Decreasing learning rate over time improves convergence
- **Subsampling Frequent Words**: Like Skip-gram, CBOW benefits from downsampling common words
- **Negative Sample Distribution**: Sampling from a modified unigram distribution ($P(w)^{0.75}$)

#### Hyperparameters

Key hyperparameters for CBOW include:

- **Window Size**: Controls how many surrounding words form the context
- **Embedding Dimension**: Typically 100-300 dimensions work well
- **Negative Samples**: 5-20 negative samples per positive example
- **Minimum Count**: Threshold for including words in vocabulary (typically 5)

````python
### CBOW Training Loop

# This is a simplified training loop for demonstration purposes
# In practice, you'd want to use DataLoader and larger datasets

def train_cbow(model, data, vocab_size, batch_size=32, epochs=5, lr=0.025):
    """Train a CBOW model on the provided dataset."""
    optimizer = optim.SGD(model.parameters(), lr=lr)
    scheduler = optim.lr_scheduler.StepLR(optimizer, step_size=1, gamma=0.9)
    
    # Prepare training batches
    num_batches = len(data) // batch_size
    losses = []
    
    for epoch in range(epochs):
        # Shuffle data for each epoch
        np.random.shuffle(data)
        epoch_loss = 0
        
        for i in range(num_batches):
            # Get batch
            batch = data[i*batch_size:(i+1)*batch_size]
            contexts_tensor, targets_tensor, neg_samples = collate_cbow_batch(batch, vocab_size)
            
            # Forward pass
            loss = model(contexts_tensor, targets_tensor, neg_samples)
            epoch_loss += loss.item()
            
            # Backward pass and optimization
            optimizer.zero_grad()
            loss.backward()
            optimizer.step()
            
            # Print progress
            if (i+1) % 100 == 0:
                print(f"Epoch {epoch+1}/{epochs}, Batch {i+1}/{num_batches}, Loss: {loss.item():.4f}")
        
        # Adjust learning rate
        scheduler.step()
        
        avg_epoch_loss = epoch_loss / num_batches
        losses.append(avg_epoch_loss)
        print(f"Epoch {epoch+1} completed, Average Loss: {avg_epoch_loss:.4f}")
    
    return losses

# Note: We don't run the training here as it would require more data
print("CBOW training function defined. In practice, you would need a larger dataset to train effectively.")
print("CBOW typically converges faster than Skip-gram but requires careful handling of variable-length contexts.")
````

### Aside: Predict or Be Predicted

The philosophical difference between CBOW and Skip-gram reflects a fundamental duality in language: Should a word predict its context, or should context predict a word? 

Skip-gram treats words as predictors of their surroundings, while CBOW treats them as targets to be predicted from context. This duality mirrors debates in linguistics about whether meaning is inherent or contextual.

Skip-gram tends to handle rare words better since each occurrence generates multiple training examples. CBOW tends to produce smoother embeddings by averaging context noise.

The same duality appears in modern language models - ELMo predicts in both directions, while BERT explicitly masks and predicts words from bi-directional context, a sophisticated evolution of the CBOW idea.

When students ask me which model to use, I often say: "Skip-gram answers 'What context surrounds this word?' while CBOW answers 'What word fits in this context?' Ask yourself which question better matches your downstream task."

### 8.5 CBOW Applications and Use Cases

CBOW embeddings have proven valuable in various NLP applications, sometimes offering advantages over Skip-gram alternatives.

#### When to Choose CBOW

CBOW is generally preferred when:

2. **The dataset is very large**: CBOW's speed advantage scales with data size
3. **Frequent words are more important**: CBOW handles frequent words better

5. **Smoother word representations are desired**: The averaging effect creates more general embeddings

#### Real-world Applications

CBOW has been successfully applied in:

- **Text Classification**: Particularly for news categorization and topic detection
- **Part-of-Speech Tagging**: Leveraging the syntactic strengths of CBOW
- **Information Retrieval**: CBOW's smoother embeddings can work well for document similarity
- **Machine Translation**: Early neural MT systems used CBOW embeddings
- **Missing Word Prediction**: Directly connects to CBOW's training objective

#### Example: Filling in the Blank

CBOW's architecture naturally solves "fill-in-the-blank" tasks, since it predicts a word given surrounding context. For instance:

"The cat sat ___ the mat" → CBOW would likely predict "on"

This makes CBOW particularly suitable for applications like:  
- Auto-completion systems
- Cloze tests in language education
- Error correction in writing assistants

````python
### Example: Fill-in-the-blank with CBOW

# This is a simplified demonstration of how CBOW could be used for fill-in-the-blank tasks
import numpy as np

class SimpleWordPredictor:
    def __init__(self, embeddings, idx_to_word, word_to_idx):
        # Store word embeddings and vocabulary mappings
        self.embeddings = embeddings  # Numpy array of embeddings
        self.idx_to_word = idx_to_word  # Map from index to word
        self.word_to_idx = word_to_idx  # Map from word to index
        
    def predict_missing_word(self, context_words, top_k=5):
        """Predict the most likely word to fill in the blank given context words."""
        # Get indices for context words, skipping unknown words
        context_indices = [self.word_to_idx[w] for w in context_words 
                          if w in self.word_to_idx]
        
        if not context_indices:
            return ["[UNK]"]  # No known context words
        
        # Get embeddings for context words
        context_vectors = self.embeddings[context_indices]
        
        # Average context vectors (simulating CBOW's context representation)
        avg_context = np.mean(context_vectors, axis=0)
        
        # Calculate similarity to all word vectors
        similarities = np.dot(self.embeddings, avg_context) / \
                      (np.linalg.norm(self.embeddings, axis=1) * np.linalg.norm(avg_context))
        
        # Get top-k most similar words
        top_indices = np.argsort(similarities)[-top_k:][::-1]
        top_words = [self.idx_to_word[idx] for idx in top_indices]
        
        return top_words

# Mock embeddings for demonstration
mock_vocab = {'the': 0, 'cat': 1, 'sat': 2, 'on': 3, 'mat': 4, 'dog': 5, 'chased': 6, 'ran': 7, 'over': 8}
mock_idx_to_word = {idx: word for word, idx in mock_vocab.items()}

# Create mock embeddings with some meaningful relationships
np.random.seed(42)  # For reproducibility
mock_embeddings = np.random.randn(len(mock_vocab), 5)  # 5-dim embeddings
# Make 'on' and 'over' similar (as prepositions)
mock_embeddings[mock_vocab['on']] = np.array([0.1, 0.2, 0.8, -0.1, 0.5])
mock_embeddings[mock_vocab['over']] = np.array([0.15, 0.25, 0.75, -0.05, 0.55])
# Make 'cat' and 'dog' somewhat similar (as animals)
mock_embeddings[mock_vocab['cat']] = np.array([0.9, -0.2, 0.1, 0.3, -0.4])
mock_embeddings[mock_vocab['dog']] = np.array([0.85, -0.25, 0.15, 0.35, -0.45])

# Create predictor
predictor = SimpleWordPredictor(mock_embeddings, mock_idx_to_word, mock_vocab)

# Test with a simple fill-in-the-blank example
context = ['the', 'cat', 'sat', 'the', 'mat']
predictions = predictor.predict_missing_word(context, top_k=2)
print(f"Context: '{' '.join(context)}' with a blank")
print(f"Top predictions to fill the blank: {predictions}")

# Try another example
context2 = ['the', 'dog', 'ran', 'the', 'mat']
predictions2 = predictor.predict_missing_word(context2, top_k=2)
print(f"Context: '{' '.join(context2)}' with a blank")
print(f"Top predictions to fill the blank: {predictions2}")
````

### 9.1 Count-Based vs. Prediction-Based Embeddings

Before diving into GloVe, let's understand the two main paradigms of word embedding methods that existed prior to its development:

#### Count-Based Methods

Count-based methods derive word representations from global corpus statistics, typically in the form of word-context co-occurrence counts. Examples include:

- **Latent Semantic Analysis (LSA)**: Applies singular value decomposition (SVD) to a term-document matrix
- **Hyperspace Analogue to Language (HAL)**: Uses a word-word co-occurrence matrix with a sliding context window
- **PPMI Matrix**: Applies positive pointwise mutual information weighting to co-occurrence counts

**Key characteristics:**
- Utilize global statistical information across the entire corpus
- Often involve matrix factorization techniques
- Computationally efficient for small-to-medium corpora
- Strong theoretical foundation in distributional semantics

#### Prediction-Based Methods

Prediction-based methods learn word representations by training neural networks to predict words from their contexts or vice versa. Examples include:

- **Word2Vec (Skip-gram)**: Predicts context words given a target word
- **Word2Vec (CBOW)**: Predicts a target word given its context words
- **Neural Language Models**: Predict the next word given a sequence of previous words

**Key characteristics:**
- Learn from local contexts rather than global statistics
- Utilize neural networks and gradient-based optimization
- Scale better to large datasets through stochastic training
- Often capture more semantic nuances and analogical relationships

For many years, these two approaches were seen as distinct and competing paradigms, with researchers debating their relative merits. However, theoretical work eventually revealed that they were more connected than initially thought.

The diagram above illustrates the conceptual difference between count-based methods (which work with global corpus statistics) and prediction-based methods (which learn from local context windows), with GloVe positioned as a bridge between these approaches.

### 9.2 GloVe: Combining Global and Local Context

GloVe (Global Vectors for Word Representation), developed by researchers at Stanford University in 2014, represents a breakthrough in word embedding technology by elegantly combining the strengths of both count-based and prediction-based approaches.

#### The Key Insight

GloVe's fundamental insight is that the ratio of co-occurrence probabilities can encode meaningful semantic relationships between words. For example, consider the words "ice" and "steam" in relation to the words "solid" and "gas":

- "Ice" co-occurs frequently with "solid" but rarely with "gas"
- "Steam" co-occurs frequently with "gas" but rarely with "solid"

The *ratios* of these probabilities contain important information about the relationships between these words:

$$\frac{P(\text{solid}|\text{ice})}{P(\text{solid}|\text{steam})} \gg 1 \quad \text{and} \quad \frac{P(\text{gas}|\text{ice})}{P(\text{gas}|\text{steam})} \ll 1$$

These ratios capture the fact that "ice" is a solid and "steam" is a gas. GloVe is designed to encode these ratio relationships directly in the vector space.

#### How GloVe Works

At a high level, GloVe:

1. Constructs a global word-word co-occurrence matrix from the corpus
2. Defines an objective function that relates co-occurrence probabilities to vector dot products
3. Uses weighted matrix factorization to learn word vectors that preserve these relationships

Unlike Word2Vec, which indirectly learns from co-occurrence patterns through context prediction, GloVe directly optimizes for these patterns in a global manner. This approach combines the computational advantages of count-based methods with the learning power of prediction-based approaches.

#### The Hybrid Nature of GloVe

GloVe can be understood from multiple perspectives:

- As a **weighted matrix factorization** of the log-co-occurrence matrix
- As a **regression model** that predicts the logarithm of co-occurrence counts
- As a **model of ratios** of co-occurrence probabilities

This multi-faceted nature makes GloVe theoretically interesting and practically powerful for capturing word meaning.

### Aside: The Best of Both Worlds

Before GloVe, word embeddings fell into two camps: count-based methods (like LSA) that captured global statistical information, and prediction-based methods (like Word2Vec) that excelled at capturing local context patterns. GloVe's key insight was that these approaches aren't fundamentally different - they're just optimizing different objectives.

By explicitly modeling the relationship between global co-occurrence statistics, GloVe bridges this gap. It's mathematically elegant: Word2Vec implicitly factorizes a shifted PMI matrix, while traditional count-based methods explicitly factorize co-occurrence matrices. GloVe's weighted objective combines these approaches, showing that the embedding wars were largely unnecessary - we could have the best of both worlds all along.

This unification was theoretically satisfying and practically beneficial. As Levy and Goldberg demonstrated in their 2014 paper, "Neural Word Embedding as Implicit Matrix Factorization," Word2Vec's Skip-gram with negative sampling implicitly factorizes a PMI matrix shifted by a constant. GloVe made this connection explicit and optimized directly for what Word2Vec was learning implicitly.

In a sense, GloVe asked: "If we know what we're trying to learn, why not optimize for it directly?"

### 9.3 The GloVe Objective Function

The mathematical formulation of GloVe is elegant and directly connects to the linguistic intuitions we've discussed. Let's walk through the derivation of the GloVe objective function step by step.

#### Starting with Co-occurrence Probabilities

Let $X_{ij}$ represent the number of times word $j$ occurs in the context of word $i$. We can define the probability of word $j$ appearing in the context of word $i$ as:

$$P(j|i) = \frac{X_{ij}}{X_i} \quad \text{where} \quad X_i = \sum_k X_{ik}$$

The ratio of co-occurrence probabilities for two context words $j$ and $k$ given two target words $i$ and $l$ is:

$$\frac{P(j|i)}{P(k|i)} / \frac{P(j|l)}{P(k|l)} = \frac{X_{ij}}{X_{ik}} \cdot \frac{X_{lk}}{X_{lj}}$$

#### From Ratios to Vector Differences

GloVe aims to capture these ratios in the vector space. The authors show that if we want the dot product of word vectors to represent log co-occurrence probabilities, we can express this as:

$$w_i^T \tilde{w}_j + b_i + \tilde{b}_j = \log(X_{ij})$$

Where:
- $w_i$ is the word vector for word $i$
- $\tilde{w}_j$ is the context vector for word $j$
- $b_i$ and $\tilde{b}_j$ are bias terms

#### The Weighted Least Squares Objective

To learn these vectors, GloVe minimizes the following weighted least squares objective:

$$J = \sum_{i,j=1}^{V} f(X_{ij}) (w_i^T \tilde{w}_j + b_i + \tilde{b}_j - \log X_{ij})^2$$

Where $f(X_{ij})$ is a weighting function defined as:

$$f(x) = \begin{cases}
(x/x_{\max})^\alpha & \text{if } x < x_{\max} \\
1 & \text{otherwise}
\end{cases}$$

With typical values of $\alpha = 3/4$ and $x_{\max} = 100$.

This weighting function serves several important purposes:

1. It prevents rare co-occurrences from being weighted the same as frequent ones
2. It stops very common co-occurrences from dominating the optimization
3. It allows zero co-occurrences to be ignored (since $f(0) = 0$)

#### Final Word Vectors

After training, GloVe produces two sets of vectors: word vectors $W$ and context vectors $\tilde{W}$. The final word embeddings are often taken as the sum or average of these two vectors:

$$\text{final_vector}_i = w_i + \tilde{w}_i$$

This averaging tends to improve performance on downstream tasks.

### 9.4 Co-occurrence Matrix Construction

The foundation of GloVe is the word-word co-occurrence matrix, which captures how often each word appears in the context of every other word. Building this matrix efficiently and appropriately is crucial for the quality of the resulting embeddings.

#### Definition

For a vocabulary of size $V$, the co-occurrence matrix $X$ is a $V \times V$ matrix where each element $X_{ij}$ represents how many times word $j$ occurs in the context of word $i$.

#### Context Window

Similar to Word2Vec, GloVe uses a sliding context window around each target word. The key parameters include:

- **Window size**: Typically 5-10 words to either side of the target word
- **Window weighting**: GloVe often uses a decreasing weighting based on distance from the target word

For example, with a window size of 5 and a harmonic weighting scheme, a context word 1 position away contributes a count of 1, a word 2 positions away contributes 1/2, and so on.

#### Efficient Construction

The co-occurrence matrix can be extremely large (vocabulary_size² elements), but is typically very sparse (most word pairs never co-occur). Efficient implementation uses sparse matrix data structures and parallelized processing to handle large corpora.

#### Example

Consider the sentence: "The cat sat on the mat"

With a window size of 2 and no distance weighting, the co-occurrence counts would include:

- ("cat", "the"): 1 (from "The cat")
- ("cat", "sat"): 1
- ("sat", "cat"): 1
- ("sat", "on"): 1
- ("on", "sat"): 1
- ("on", "the"): 1 (from "on the")
- etc.

With distance weighting, the counts would be adjusted by the distance between words.

````python
### Code Example: Co-occurrence Matrix Construction

import numpy as np
from collections import defaultdict, Counter
import scipy.sparse as sparse

def build_cooccurrence_matrix(corpus, vocab_size, window_size=5, min_count=5):
    """Build word co-occurrence matrix from a corpus.
    
    Args:
        corpus: List of tokenized sentences (each a list of words)
        vocab_size: Size of vocabulary to use
        window_size: Context window size (on each side)
        min_count: Minimum word count to include in vocab
        
    Returns:
        cooccurrence matrix, word-to-index mapping
    """
    # Build vocabulary
    word_counts = Counter([word for sent in corpus for word in sent])
    # Filter by minimum count and take most common
    vocab = [word for word, count in word_counts.most_common(vocab_size) 
             if count >= min_count]
    word_to_id = {word: i for i, word in enumerate(vocab)}
    
    # Initialize co-occurrence dictionary
    cooccurrences = defaultdict(float)
    
    # Build co-occurrence matrix
    for sentence in corpus:
        # Convert to word IDs, skipping OOV words
        word_ids = [word_to_id[w] for w in sentence if w in word_to_id]
        for center_i, center_id in enumerate(word_ids):
            # For each context position
            for context_i in range(max(0, center_i - window_size), 
                               min(len(word_ids), center_i + window_size + 1)):
                if context_i != center_i:  # Skip the word itself
                    context_id = word_ids[context_i]
                    # Weight by inverse of distance
                    distance = abs(center_i - context_i)
                    cooccurrences[(center_id, context_id)] += 1.0 / distance
    
    # Convert to sparse matrix
    i_indices = [i for (i, j), _ in cooccurrences.items()]
    j_indices = [j for (i, j), _ in cooccurrences.items()]
    values = [count for _, count in cooccurrences.items()]
    
    cooc_matrix = sparse.coo_matrix(
        (values, (i_indices, j_indices)), 
        shape=(len(vocab), len(vocab))
    )
    
    return cooc_matrix, word_to_id

# Example usage
corpus = [
    "the cat sat on the mat".split(),
    "the dog played in the yard".split(),
    "the cat and the dog are friends".split()
]

cooc_matrix, word_to_id = build_cooccurrence_matrix(corpus, vocab_size=10, window_size=2)
print(f"Vocabulary: {word_to_id}")
print(f"Co-occurrence matrix shape: {cooc_matrix.shape}")
print(f"Number of non-zero entries: {cooc_matrix.nnz}")

# Show a few entries
cooc_matrix = cooc_matrix.tocsr()  # Convert to CSR for efficient element access
for word1 in ['the', 'cat', 'dog']:
    if word1 in word_to_id:
        i = word_to_id[word1]
        for word2 in ['the', 'cat', 'dog', 'sat', 'on']:
            if word2 in word_to_id:
                j = word_to_id[word2]
                print(f"Co-occurrence ({word1}, {word2}): {cooc_matrix[i, j]}")
````

### Aside: The Matrix View

Understanding embeddings as matrix factorization provides powerful insights. In this view, word embedding methods decompose a large matrix (either explicit like GloVe's co-occurrence matrix or implicit like Word2Vec's shifted PMI matrix) into two lower-dimensional matrices. The rows of one matrix become word embeddings.

This connects embeddings to classical techniques like Principal Component Analysis and helps explain why they work: they're capturing the main dimensions of variation in word contexts. The matrix perspective also inspired later work on embeddings for graphs, knowledge bases, and recommendations - all fundamentally matrix factorization problems.

Some researchers argue this view demystifies the 'magic' of embeddings, showing they're principled approximations of linguistic statistics rather than black-box neural networks.

When we view GloVe through this lens, we see that it's factorizing a log-transformed co-occurrence matrix, where each element $M_{ij} = \log(X_{ij})$ (with some additional weighting and bias terms). The resulting factors $W$ and $\tilde{W}$ are our word and context embeddings.

This perspective unifies various embedding approaches:
- LSA factorizes a term-document matrix
- PPMI+SVD factorizes a positive pointwise mutual information matrix
- Word2Vec implicitly factorizes a shifted PMI matrix
- GloVe explicitly factorizes a weighted log co-occurrence matrix

The matrix view reminds us that these methods, despite their different motivations and implementations, are fundamentally doing similar things mathematically.

### 9.5 Training and Using GloVe Embeddings

Once we've constructed the co-occurrence matrix, we can train GloVe embeddings by optimizing the weighted least squares objective function described earlier.

#### Training Process

The training process for GloVe typically involves:

1. **Initialization**: Randomly initialize word vectors $W$ and context vectors $\tilde{W}$, along with bias terms
2. **Optimization**: Use stochastic gradient descent or Adagrad to minimize the objective function
3. **Iteration**: Train for a fixed number of epochs (typically 50-100) or until convergence
4. **Combination**: Add or average the word and context vectors to get the final embeddings

#### Hyperparameters

Key hyperparameters for GloVe include:

- **Vector dimension**: Typically 50-300 dimensions
- **Window size**: Size of context window used to build co-occurrence matrix (5-10 is common)
- **Minimum count**: Threshold for including words in the vocabulary
- **Learning rate**: Step size for gradient updates (0.05 is common with Adagrad)
- **Weighting parameters**: $\alpha$ (typically 0.75) and $x_{max}$ (typically 100) for the weighting function

#### Pre-trained Embeddings

The Stanford NLP group provides pre-trained GloVe embeddings trained on several large corpora:

- Wikipedia 2014 + Gigaword 5 (6B tokens)
- Common Crawl (42B tokens)
- Common Crawl (840B tokens)
- Twitter (27B tokens)

These embeddings are widely used as feature inputs for downstream NLP tasks.

#### Using GloVe Embeddings

Like Word2Vec embeddings, GloVe embeddings can be used for a variety of NLP tasks:

- As input features for neural networks
- For semantic similarity and analogy tasks
- For document representation (e.g., by averaging word vectors)
- As initialization for task-specific embeddings

````python
### Code Example: Training Simple GloVe Vectors

import numpy as np
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import Dataset, DataLoader

class GloVeDataset(Dataset):
    def __init__(self, cooc_matrix):
        # Convert sparse matrix to COO format for easy iteration
        self.cooc_matrix = cooc_matrix.tocoo()
        self.i_indices = torch.LongTensor(self.cooc_matrix.row)
        self.j_indices = torch.LongTensor(self.cooc_matrix.col)
        self.values = torch.FloatTensor(self.cooc_matrix.data)
    
    def __len__(self):
        return len(self.values)
    
    def __getitem__(self, idx):
        return self.i_indices[idx], self.j_indices[idx], self.values[idx]

class GloVeModel(nn.Module):
    def __init__(self, vocab_size, embedding_dim):
        super(GloVeModel, self).__init__()
        
        # Word embeddings
        self.w_embeddings = nn.Embedding(vocab_size, embedding_dim)
        # Context embeddings
        self.c_embeddings = nn.Embedding(vocab_size, embedding_dim)
        # Bias terms
        self.w_biases = nn.Embedding(vocab_size, 1)
        self.c_biases = nn.Embedding(vocab_size, 1)
        
        # Initialize embeddings
        self.w_embeddings.weight.data.uniform_(-0.5/embedding_dim, 0.5/embedding_dim)
        self.c_embeddings.weight.data.uniform_(-0.5/embedding_dim, 0.5/embedding_dim)
        self.w_biases.weight.data.zero_()
        self.c_biases.weight.data.zero_()
    
    def forward(self, i_indices, j_indices):
        w_embeds = self.w_embeddings(i_indices)  # [batch_size, embedding_dim]
        c_embeds = self.c_embeddings(j_indices)  # [batch_size, embedding_dim]
        w_biases = self.w_biases(i_indices).squeeze()  # [batch_size]
        c_biases = self.c_biases(j_indices).squeeze()  # [batch_size]
        
        # w_i^T c_j + b_i + c_j
        log_predictions = torch.sum(w_embeds * c_embeds, dim=1) + w_biases + c_biases
        
        return log_predictions
    
    def get_embeddings(self):
        # Get final word embeddings as sum of word and context vectors
        return self.w_embeddings.weight.data + self.c_embeddings.weight.data

def train_glove(cooc_matrix, vocab_size, embedding_dim=100, epochs=50, batch_size=512, lr=0.05):
    """Train GloVe embeddings from a co-occurrence matrix."""
    device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
    
    # Create dataset and dataloader
    dataset = GloVeDataset(cooc_matrix)
    dataloader = DataLoader(dataset, batch_size=batch_size, shuffle=True)
    
    # Initialize model
    model = GloVeModel(vocab_size, embedding_dim).to(device)
    optimizer = optim.Adagrad(model.parameters(), lr=lr)
    
    # GloVe weighting function
    def weight_func(x, x_max=100, alpha=0.75):
        return torch.minimum(torch.pow(x / x_max, alpha), torch.tensor(1.0))
    
    # Training loop
    model.train()
    for epoch in range(epochs):
        total_loss = 0
        for i_indices, j_indices, counts in dataloader:
            i_indices = i_indices.to(device)
            j_indices = j_indices.to(device)
            counts = counts.to(device)
            
            # Forward pass
            log_predictions = model(i_indices, j_indices)
            log_counts = torch.log(counts)
            
            # Compute weights
            weights = weight_func(counts)
            
            # Weighted loss
            loss = weights * torch.pow(log_predictions - log_counts, 2)
            loss = loss.mean()
            
            # Backward pass and optimize
            optimizer.zero_grad()
            loss.backward()
            optimizer.step()
            
            total_loss += loss.item()
            
        # Print progress
        if (epoch + 1) % 5 == 0:
            print(f'Epoch {epoch+1}/{epochs}, Loss: {total_loss/len(dataloader):.6f}')
    
    # Return final embeddings
    return model.get_embeddings().cpu().numpy()

# We're not actually running this heavy training here since it would be too slow
# in a notebook environment. In practice, you would run:
# embeddings = train_glove(cooc_matrix, len(word_to_id), embedding_dim=50, epochs=50)

print("GloVe training code ready! (Not executed to save computation)")

# Using pre-trained GloVe embeddings
import os
import requests
from zipfile import ZipFile
import io

def load_pretrained_glove(dimension=50):
    """Load pre-trained GloVe vectors."""
    # This is just an example of how you would load pre-trained vectors
    # In an actual notebook, you'd download from Stanford's website
    print(f"Downloading pre-trained GloVe embeddings ({dimension}d)...")
    print("(In a real notebook, this would download actual files)")
    
    # Normally we'd do something like:
    # url = f"http://nlp.stanford.edu/data/glove.6B.{dimension}d.zip"
    # response = requests.get(url)
    # with ZipFile(io.BytesIO(response.content)) as zip_file:
    #     with zip_file.open(f"glove.6B.{dimension}d.txt") as f:
    #         # Process the file
    
    # For this example, we'll just create some dummy embeddings
    word_to_vec = {}
    word_to_vec["cat"] = np.random.randn(dimension)
    word_to_vec["dog"] = np.random.randn(dimension)
    word_to_vec["king"] = np.random.randn(dimension)
    word_to_vec["queen"] = np.random.randn(dimension)
    word_to_vec["man"] = np.random.randn(dimension)
    word_to_vec["woman"] = np.random.randn(dimension)
    
    return word_to_vec

# Example: Load pre-trained GloVe embeddings
glove_vectors = load_pretrained_glove(dimension=50)
print(f"Loaded {len(glove_vectors)} pre-trained GloVe vectors with dimension {len(next(iter(glove_vectors.values())))}")

# Example: Calculate similarity between words
def cosine_similarity(vec1, vec2):
    return np.dot(vec1, vec2) / (np.linalg.norm(vec1) * np.linalg.norm(vec2))

print(f"Similarity between 'man' and 'woman': {cosine_similarity(glove_vectors['man'], glove_vectors['woman']):.4f}")
print(f"Similarity between 'king' and 'queen': {cosine_similarity(glove_vectors['king'], glove_vectors['queen']):.4f}")
print(f"Similarity between 'cat' and 'dog': {cosine_similarity(glove_vectors['cat'], glove_vectors['dog']):.4f}")
````

### 10.1 Intrinsic Evaluation Methods

Intrinsic evaluation methods assess embeddings directly by testing how well they capture semantic and syntactic relationships between words. These methods don't require integration into a downstream task, making them quick and straightforward to apply.

#### Word Similarity Tasks

In word similarity tasks, we compare the cosine similarity between word pairs in the embedding space with human judgments of semantic similarity. Standard datasets include:

- **WordSim-353**: 353 word pairs with human similarity judgments
- **SimLex-999**: 999 word pairs with explicit focus on similarity rather than relatedness
- **RG-65**: One of the earliest datasets with 65 noun pairs
- **MEN**: 3,000 word pairs with similarity ratings

Performance is typically measured using Spearman's rank correlation coefficient between human judgments and embedding similarities.

#### Word Analogy Tasks

Word analogy tasks test the embedding's ability to capture relational similarities, based on the observation that vector differences can encode semantic and syntactic relationships.

The classic example is: `king - man + woman ≈ queen`

Standard datasets include:

- **Google Analogy Dataset**: 19,544 questions across semantic and syntactic categories
- **BATS**: Balanced Analogy Test Set with more comprehensive coverage of relation types

Accuracy is measured as the percentage of questions where the correct answer is the closest word vector to the computed target (excluding the input words).

### Aside: The Benchmark Problem

Evaluating embeddings presents a fascinating methodological challenge in NLP research. The field has produced dozens of benchmarks, but comparing results across them can lead to contradictory conclusions about which embedding method is "best."

For example, a 2015 study found that Word2Vec performed better on analogy tasks while GloVe excelled at word similarity benchmarks. Later, when evaluated on specialized domains like biomedicine, domain-specific embeddings outperformed both, despite being trained on much less data.

This benchmark inconsistency reveals a deeper truth: there is no universally "best" embedding. Different methods capture different aspects of meaning, and their performance depends heavily on:

1. The linguistic properties being tested
2. Domain specificity of the evaluation data
3. Size and quality of the training corpus
4. Hyperparameter settings (window size, dimension, etc.)

What's more concerning is that researchers have found that performance on intrinsic tasks like word similarity often correlates poorly with performance on downstream applications. An embedding might excel at word analogies but perform poorly when used in a sentiment classifier.

This has led to a pragmatic approach: instead of chasing state-of-the-art on benchmarks, practitioners often try several embedding options and select whatever works best for their specific task. The field continues to develop more comprehensive evaluation frameworks that better predict real-world utility rather than abstract linguistic properties.

### 10.2 Extrinsic Evaluation on Downstream Tasks

While intrinsic evaluations test linguistic properties directly, extrinsic evaluations measure how much embeddings improve performance on actual NLP applications. These evaluations better reflect real-world utility but are more resource-intensive to conduct.

#### Common Downstream Tasks

1. **Text Classification**
   - Sentiment analysis
   - Topic categorization
   - Intent detection

2. **Named Entity Recognition (NER)**
   - Identifying and classifying named entities (people, organizations, locations, etc.)

3. **Part-of-Speech (POS) Tagging**
   - Assigning grammatical categories to words

4. **Parsing**
   - Determining the syntactic structure of sentences

5. **Machine Translation**
   - Converting text from one language to another

#### Evaluation Methodology

A typical extrinsic evaluation follows these steps:

1. Create multiple versions of the same model architecture
2. Vary only the embedding method between versions
3. Train each model on the same task data
4. Measure performance differences using task-specific metrics

![Downstream Task Performance](https://miro.medium.com/v2/resize:fit:2000/1*fTftp5SQ3tSjagJ1pe65Bg.png)

Importantly, different tasks may benefit from different embedding properties. For example:

- Syntactic tasks like POS tagging benefit from embeddings that capture grammatical relationships
- Semantic tasks like sentiment analysis benefit from embeddings that capture meaning and connotation

### 10.3 Visualization Techniques

Visualizing word embeddings helps us understand what they've learned and identify patterns or problems that aren't obvious from quantitative evaluations. Since embeddings typically have hundreds of dimensions, visualization requires dimensionality reduction techniques.

#### t-SNE: t-Distributed Stochastic Neighbor Embedding

t-SNE is the most popular technique for embedding visualization. It works by:

1. Computing pairwise similarities between points in the high-dimensional space
2. Finding a low-dimensional representation (typically 2D or 3D) that preserves these similarities
3. Emphasizing the preservation of local structure over global structure

t-SNE is particularly good at revealing clusters of semantically related words and has produced many of the iconic visualizations of word embeddings that helped popularize the technology.

#### UMAP: Uniform Manifold Approximation and Projection

UMAP is a newer alternative to t-SNE that:

1. Often better preserves global structure
2. Has faster computation time
3. Provides more stable results across runs

#### PCA: Principal Component Analysis

While simpler than t-SNE or UMAP, PCA can still provide valuable visualizations by:

1. Finding the principal axes of variation in the embedding space
2. Projecting embeddings onto these axes
3. Preserving global variance at the expense of local structure

![Embedding Visualization Comparison](https://cdn-images-1.medium.com/max/1600/1*QbSwdnm6JLwJsqpSaTEung.png)

#### Interpreting Visualizations

When examining embedding visualizations, look for:

1. **Semantic clusters**: Words with similar meanings should appear close together
2. **Syntactic patterns**: Words with similar grammatical functions may form distinct regions
3. **Analogical relationships**: Some relationships may appear as parallel lines or regular geometric patterns
4. **Outliers and anomalies**: Words that appear far from similar terms may indicate problems
5. **Bias indicators**: Problematic associations (e.g., gendered occupation clusters) may be visible

Remember that visualizations can be misleading - they necessarily distort some relationships to display others. Different parameter settings in t-SNE or UMAP can produce very different visualizations of the same embeddings.

### 10.4 Specialized Embeddings for Specific Domains

General-purpose embeddings like GloVe or Word2Vec are trained on broad corpora like Wikipedia, news articles, or web text. While they capture common language patterns, they often fall short for specialized domains with unique vocabularies and semantic relationships.

#### Domain-Specific Challenges

Specialized domains present several challenges for word embeddings:

1. **Technical terminology**: Words with domain-specific meanings
2. **Rare words**: Important terms that appear infrequently in general corpora
3. **Semantic shifts**: Common words with different meanings in the domain
4. **Abbreviations and acronyms**: Short forms prevalent in technical fields

#### Approaches for Domain Adaptation

1. **Train from scratch**: Train embeddings exclusively on domain-specific corpora
   - *Advantages*: Captures domain semantics accurately
   - *Disadvantages*: Requires substantial domain text; loses general language knowledge

2. **Fine-tuning**: Start with pre-trained general embeddings and continue training on domain text
   - *Advantages*: Combines general and domain-specific knowledge
   - *Disadvantages*: May catastrophically forget general patterns

3. **Retrofitting**: Modify pre-trained embeddings to respect domain-specific constraints
   - *Advantages*: Preserves most of the original structure
   - *Disadvantages*: Limited ability to inject new domain knowledge

4. **Hybrid approaches**: Combine multiple embedding sources
   - *Advantages*: Flexible integration of different knowledge sources
   - *Disadvantages*: Complex to implement and tune

#### Domain Examples

**Biomedical**
Biomedical texts contain highly specialized terminology. BioWordVec, Bio_ClinicalBERT, and other domain-specific embeddings outperform general embeddings on tasks like disease name recognition and clinical concept extraction.

**Legal**
Legal language uses ordinary words with specialized meanings and complex technical terms. Legal-BERT and similar models capture these nuances better than general models.

**Financial**
Financial texts include company names, technical indicators, and terminology with specific meanings. FinBERT and similar models show superior performance for sentiment analysis of financial news and reports.

### 10.5 From Static to Contextual Embeddings

All the embedding methods we've discussed so far share a fundamental limitation: they assign the same vector to a word regardless of context. This static representation fails to capture polysemy (words with multiple meanings) and contextual nuances.

#### The Polysemy Problem

Consider the word "bank":
- "I deposited money in the bank" (financial institution)
- "We sat on the river bank" (shore)

In Word2Vec, GloVe, or FastText, "bank" receives a single vector that awkwardly averages these distinct meanings.

#### The Contextual Revolution

Contextual embeddings address this limitation by generating different representations for the same word depending on its surrounding context. Key developments include:

1. **CoVe (2017)**: Contextualized Word Vectors derived from an MT encoder

2. **ELMo (2018)**: Embeddings from Language Models using bidirectional LSTM
   - Pre-trains a bi-directional language model
   - Uses all layers of the LSTM to create contextualized representations
   - Significantly improved performance on multiple NLP benchmarks

3. **BERT (2018)**: Bidirectional Encoder Representations from Transformers
   - Uses transformer architecture instead of LSTM
   - Trains with masked language modeling objective
   - Creates deeply contextualized representations

4. **GPT, RoBERTa, T5, etc.**: Further refined contextual embedding approaches

![Static vs Contextual Embeddings](https://www.researchgate.net/publication/348443063/figure/fig2/AS:979873639010307@1610644722794/A-comparison-between-static-embeddings-and-contextual-embeddings-The-two-instances-of.png)

#### Key Advantages of Contextual Embeddings

1. **Sense disambiguation**: Different vectors for different word senses
2. **Contextual nuance**: Capture subtle meaning variations based on context
3. **Dynamic representation**: Adapt to new contexts without retraining
4. **Improved downstream performance**: Significant gains across NLP tasks

#### Relationship to Static Embeddings

Contextual embeddings haven't replaced static embeddings but represent an evolution:

1. The conceptual foundations remain similar (distributional hypothesis)
2. Static embeddings are still useful for simpler tasks and limited resources
3. Many contextual models still use static embeddings in their first layer
4. Understanding static embeddings provides crucial background for working with contextual models

As we'll explore in later sections, contextual embeddings form the foundation of modern transformer-based language models that have revolutionized NLP.

### 11.2 The Evolution of Text Representation

The journey of text representation in NLP reflects our growing understanding of language and the increasing capabilities of our models:

#### Early Days: Symbolic Approaches
- **Boolean bag-of-words**: Simple presence/absence of words
- **TF-IDF**: Weighting words by frequency and document distinctiveness
- **One-hot encodings**: Representing each word as a sparse vector with a single 1

These approaches were computationally simple but failed to capture semantic relationships. A major breakthrough came with the realization that we could represent words as dense vectors in a continuous space.

#### The Static Embedding Era
- **Neural language models** (2003): Bengio et al. introduced neural networks for language modeling with word embeddings as a byproduct
- **Word2Vec** (2013): Mikolov et al. introduced efficient algorithms for learning word vectors, popularizing the approach
- **GloVe** (2014): Pennington et al. combined global statistics and local context in a unified model
- **FastText** (2016): Bojanowski et al. extended Word2Vec to use character n-grams, better handling morphology

Static embeddings revolutionized NLP by capturing semantic and syntactic relationships in dense vectors. However, they suffered from key limitations, particularly their inability to handle polysemy (words with multiple meanings).

#### The Contextual Revolution
- **ELMo** (2018): Peters et al. introduced contextualized word representations from bidirectional LSTMs
- **BERT** (2018): Devlin et al. used transformer architectures to generate deeply contextual representations
- **GPT models** (2018-present): OpenAI's series of increasingly powerful autoregressive language models

These contextual models generate different representations for the same word depending on its context, addressing the polysemy problem and capturing much richer semantic information.

Throughout this evolution, we can trace a clear trend: moving from sparse, discrete representations to dense, continuous vectors, and from static to dynamic, contextualized representations.

````python
# A simple end-to-end NLP pipeline integrating concepts from throughout the notebook

import torch
import torch.nn as nn
import torch.optim as optim
from torchtext.vocab import GloVe
from collections import Counter
import re
import numpy as np
from sklearn.metrics import accuracy_score
import matplotlib.pyplot as plt
from sklearn.manifold import TSNE

# Sample data (in practice, you'd use a real dataset)
train_texts = [
    "I loved this movie, it was fantastic!",
    "The acting was terrific and the plot kept me engaged",
    "What a waste of time, terrible movie",
    "I hated every minute of this film",
    "Brilliant performance by the entire cast",
    "This was the worst movie I've seen all year",
    "Absolutely enjoyed this masterpiece",
    "Boring plot and wooden acting"
]
train_labels = [1, 1, 0, 0, 1, 0, 1, 0]  # 1=positive, 0=negative

test_texts = [
    "I really enjoyed this movie",
    "The worst film ever made",
    "A delightful cinematic experience",
    "Don't waste your money on this garbage"
]
test_labels = [1, 0, 1, 0]

# 1. Tokenization (word-level for simplicity)
def tokenize(text):
    # Simple word tokenization with lowercasing and punctuation removal
    return re.findall(r'\w+', text.lower())

# Apply tokenization
train_tokens = [tokenize(text) for text in train_texts]
test_tokens = [tokenize(text) for text in test_texts]

# 2. Build vocabulary from training data
all_train_tokens = [token for tokens in train_tokens for token in tokens]
counter = Counter(all_train_tokens)
vocab = ['<pad>', '<unk>'] + [token for token, count in counter.most_common() if count > 1]
token2idx = {token: idx for idx, token in enumerate(vocab)}

# 3. Convert tokens to indices
def tokens_to_indices(tokens, token2idx):
    return [token2idx.get(token, token2idx['<unk>']) for token in tokens]

train_indices = [tokens_to_indices(tokens, token2idx) for tokens in train_tokens]
test_indices = [tokens_to_indices(tokens, token2idx) for tokens in test_tokens]

# 4. Pad sequences to same length
max_len = max(len(indices) for indices in train_indices + test_indices)

def pad_sequences(sequences, max_len):
    return [seq + [0] * (max_len - len(seq)) for seq in sequences]

train_padded = pad_sequences(train_indices, max_len)
test_padded = pad_sequences(test_indices, max_len)

# 5. Load pre-trained GloVe embeddings
# Note: This requires downloading GloVe embeddings - for a notebook we'll simulate them
embedding_dim = 50
# In practice: glove = GloVe(name='6B', dim=50)

# Simulating GloVe for demonstration
np.random.seed(42)
glove_vectors = {}
for token in vocab:
    if token == '<pad>':
        glove_vectors[token] = np.zeros(embedding_dim)
    else:
        glove_vectors[token] = np.random.randn(embedding_dim)

# Create embedding matrix
embedding_matrix = np.zeros((len(vocab), embedding_dim))
for i, token in enumerate(vocab):
    embedding_matrix[i] = glove_vectors.get(token, np.random.randn(embedding_dim) * 0.1)

# 6. Define the model
class SimpleClassifier(nn.Module):
    def __init__(self, vocab_size, embedding_dim, hidden_dim, output_dim, embedding_matrix=None):
        super(SimpleClassifier, self).__init__()
        
        self.embedding = nn.Embedding(vocab_size, embedding_dim)
        if embedding_matrix is not None:
            self.embedding.weight = nn.Parameter(torch.FloatTensor(embedding_matrix))
            
        self.fc1 = nn.Linear(embedding_dim, hidden_dim)
        self.fc2 = nn.Linear(hidden_dim, output_dim)
        self.dropout = nn.Dropout(0.3)
        
    def forward(self, text):
        # text = [batch size, sent len]
        embedded = self.embedding(text)  # [batch size, sent len, emb dim]
        
        # Simple average pooling of word embeddings
        pooled = torch.mean(embedded, dim=1)  # [batch size, emb dim]
        
        hidden = torch.relu(self.fc1(pooled))
        hidden = self.dropout(hidden)
        output = self.fc2(hidden)
        
        return output

# Convert data to PyTorch tensors
X_train = torch.LongTensor(train_padded)
y_train = torch.FloatTensor(train_labels)
X_test = torch.LongTensor(test_padded)
y_test = torch.FloatTensor(test_labels)

# Initialize model and training parameters
model = SimpleClassifier(
    vocab_size=len(vocab),
    embedding_dim=embedding_dim,
    hidden_dim=32,
    output_dim=1,
    embedding_matrix=embedding_matrix
)

optimizer = optim.Adam(model.parameters())
criterion = nn.BCEWithLogitsLoss()

# 7. Train the model
def train_model(model, X_train, y_train, epochs=20):
    model.train()
    for epoch in range(epochs):
        optimizer.zero_grad()
        predictions = model(X_train).squeeze()
        loss = criterion(predictions, y_train)
        loss.backward()
        optimizer.step()
        
        # Print progress every few epochs
        if (epoch+1) % 5 == 0:
            print(f'Epoch {epoch+1}/{epochs}, Loss: {loss.item():.4f}')

# 8. Evaluate the model
def evaluate_model(model, X_test, y_test):
    model.eval()
    with torch.no_grad():
        predictions = model(X_test).squeeze()
        predicted_labels = (torch.sigmoid(predictions) > 0.5).float()
        accuracy = accuracy_score(y_test, predicted_labels)
        print(f'Test accuracy: {accuracy:.4f}')
        
        # Show predictions
        for i, (text, true, pred) in enumerate(zip(test_texts, y_test, predicted_labels)):
            sentiment = "positive" if pred.item() == 1 else "negative"
            correct = "✓" if true == pred.item() else "✗"
            print(f"{correct} \"{text}\" → {sentiment}")
            
# 9. Visualize word embeddings
def visualize_embeddings(model, vocab):
    # Get word embeddings from the model
    embeddings = model.embedding.weight.detach().numpy()
    
    # Choose interesting words to visualize
    sentiment_words = ['loved', 'enjoyed', 'brilliant', 'waste', 'terrible', 'worst', 'hated', 'boring']
    sentiment_indices = [token2idx.get(word, token2idx['<unk>']) for word in sentiment_words]
    
    # Get embeddings for these words
    sentiment_embeddings = embeddings[sentiment_indices]
    
    # Apply t-SNE to reduce dimensionality for visualization
    tsne = TSNE(n_components=2, random_state=42)
    sentiment_embeddings_2d = tsne.fit_transform(sentiment_embeddings)
    
    # Plot the embeddings
    plt.figure(figsize=(10, 8))
    plt.scatter(sentiment_embeddings_2d[:, 0], sentiment_embeddings_2d[:, 1], c=['g', 'g', 'g', 'r', 'r', 'r', 'r', 'r'])
    
    # Add word labels
    for i, word in enumerate(sentiment_words):
        plt.annotate(word, (sentiment_embeddings_2d[i, 0], sentiment_embeddings_2d[i, 1]), fontsize=12)
        
    plt.title('t-SNE Visualization of Sentiment Word Embeddings')
    plt.xlabel('Dimension 1')
    plt.ylabel('Dimension 2')
    plt.show()

# Uncomment these lines to actually run the training and evaluation
# train_model(model, X_train, y_train)
# evaluate_model(model, X_test, y_test)
# visualize_embeddings(model, vocab)

print("This code demonstrates a complete pipeline incorporating tokenization, embeddings, and classification.")
````

#### Design Patterns in Text Representation

Throughout our journey, we've seen several design patterns emerge in text representation approaches:

1. **The Embedding-Prediction Loop**: Learning representations by predicting context (Skip-gram), target (CBOW), or co-occurrence statistics (GloVe)
2. **The Vocabulary Tradeoff**: Balancing coverage (character) vs. meaning (word) vs. efficiency (subword)
3. **The Static-Dynamic Spectrum**: Moving from fixed representations to context-dependent ones
4. **The Bottom-Up Construction**: Building language understanding from characters to words to sentences to documents

These patterns continue to inform modern approaches, even as architectures grow more complex. The transformer models that dominate NLP today still use tokenization (typically subword), still learn embeddings (though contextual rather than static), and still fundamentally rely on distributional patterns in text.

### Flash Cards for Notebook Review

**Card 1:**
- **Front**: What are the three main approaches to tokenization we've covered?
- **Back**: Character tokenization (each character as a token), word tokenization (each word as a token), and subword tokenization (learned fragments of words as tokens, e.g., BPE, WordPiece, SentencePiece).

**Card 2:**
- **Front**: Explain the distributional hypothesis and how it relates to word embeddings.
- **Back**: The distributional hypothesis states that words appearing in similar contexts tend to have similar meanings ("you shall know a word by the company it keeps"). Word embeddings operationalize this by learning representations where words with similar contexts have similar vectors.

**Card 3:**
- **Front**: What key limitation of static word embeddings do contextual embeddings address?
- **Back**: Static embeddings assign the same vector to a word regardless of context, failing to capture polysemy (multiple word meanings). Contextual embeddings generate different representations based on surrounding text, allowing the same word to have different vectors in different contexts.

**Card 4:**
- **Front**: How do Skip-gram and CBOW models differ in their approach to learning word embeddings?
- **Back**: Skip-gram predicts context words given a target word (one-to-many prediction), while CBOW predicts a target word given its context words (many-to-one prediction). Skip-gram often handles rare words better, while CBOW is typically faster to train.

**Card 5:**
- **Front**: How does GloVe differ from Word2Vec in its training approach?
- **Back**: GloVe uses global matrix factorization of word co-occurrence statistics, while Word2Vec uses local context window prediction. GloVe explicitly trains on co-occurrence probabilities, while Word2Vec learns them implicitly through prediction tasks.
