The high level idea of affine transformation in math is that for each column vector $\mathbf{x} \in \Bbb R^{in\_features}$, an affine transformation maps it to another column vector $\mathbf{y} \in \Bbb R^{out\_features}$ via

$$
\mathbf{y} = \mathbf{W} \mathbf{x} + \mathbf{b} ,
$$

where

* $\mathbf{W} \in \Bbb R^{out\_features \times in\_features}$.
